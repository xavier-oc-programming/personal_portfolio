# Phase 3 · Section 2 — Project Data Persistence

### Objective

Phase 3 Section 2 transitions the portfolio application from an in-memory dataset to a persistent data layer.

In earlier phases, project information was loaded from the Python dataset located in `data/projects.py`. While this approach simplified development, it did not provide true persistence or scalability.

This section introduces **database seeding**, allowing the existing project dataset to be inserted into the SQLite database defined in Phase 3 Section 1.

By the end of this section, the portfolio database contains the full project dataset, establishing the foundation for database-driven queries in later sections.

---

## Prerequisites

Before completing this section, the following components must already exist:

* SQLite database configuration
* SQLAlchemy ORM integration
* `Project` model schema
* JSON serialization strategy for list fields

These were implemented in **Phase 3 · Section 1 — Database Foundation**.

---

## Implementation Steps

The following steps are performed in this section:

1. Introduce a database seeding process.
2. Convert project dictionaries into SQLAlchemy ORM objects.
3. Insert project records into the SQLite database.
4. Verify that the database contains the expected project data.

This structured process allows the existing dataset to be migrated safely without disrupting the application's current behavior.

---

## Database Seeding

A database seeding process is introduced to populate the `projects` table with the existing project dataset.

The seeding process reads the project data from:

```
data/projects.py
```

Each project dictionary is converted into a `Project` ORM object and inserted into the SQLite database.

This approach ensures that the database schema defined in Phase 3 Section 1 is populated with the same structured data used during Phase 2.

---

## ORM Object Creation

Each project entry from the dataset is mapped to the SQLAlchemy `Project` model.

For example, a project dictionary is converted into an ORM instance:

```
Project(
    slug=project["slug"],
    title=project["title"],
    primary_category=project["primary_category"],
    short_description=project["short_description"],
    full_description=project["full_description"],
)
```

Additional attributes such as tags, tech stack, screenshots, and videos are stored using the JSON serialization strategy defined in Section 1.4.

This ensures that list-based fields remain compatible with the relational database structure.

---

## Database Insertion

Once ORM objects are created, they are added to the SQLAlchemy session and committed to the database.

Typical insertion workflow:

```
db.session.add(project_instance)
db.session.commit()
```

The database now persists project records rather than relying on temporary in-memory data.

---

## Persistence Verification

After the seeding process runs, the SQLite database contains the full set of portfolio projects.

The resulting table structure can be inspected using tools such as:

* DB Browser for SQLite
* SQLite CLI
* SQLAlchemy queries

The `projects` table now stores all portfolio entries as persistent records.

---

## Temporary Dual Data Sources

During this phase, the application **continues to render project pages from the in-memory dataset** while the database layer is populated and verified.

This staged approach ensures that:

* the database schema can be validated safely
* data integrity can be confirmed
* the application remains fully functional

Future sections will update routes to query the database instead of the Python dataset.

## Result

With Section 2 complete, the portfolio database now contains the full project dataset.

The application now includes:

* a persistent SQLite database
* a structured `projects` table populated with real data
* ORM-based project models
* a repeatable database seeding process

This establishes the first fully persistent data layer within the portfolio project and prepares the application for the next stage, where routes will begin retrieving project data directly from the database.
