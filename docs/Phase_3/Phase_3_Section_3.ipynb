# Phase 3 · Section 3 — Contact Message Storage

### Objective

Phase 3 Section 3 introduces the ability for the portfolio application to receive and store contact messages submitted through the website.

In earlier phases, the contact page functioned only as a static information page displaying contact details. While this allowed visitors to find ways to reach the portfolio owner, it did not provide a direct communication channel through the application itself.

This section introduces a **database-backed contact message system**, allowing visitors to submit inquiries through a contact form.

By the end of this section, contact form submissions are validated and stored in the SQLite database, establishing the foundation for managing incoming messages within the application.

---

## Prerequisites

Before completing this section, the following components must already exist:

* SQLite database configuration
* SQLAlchemy ORM integration
* database initialization logic
* persistent data storage infrastructure

These components were implemented in **Phase 3 · Section 1 — Database Foundation**.

---

## Implementation Steps

The following steps are performed in this section:

1. Introduce a database model for contact messages.
2. Implement server-side validation for contact form submissions.
3. Create a timestamped record for each message.
4. Store validated messages in the SQLite database.

This structured process enables the portfolio to receive and persist visitor inquiries while maintaining clean input validation rules.

---

## Contact Message Model

A new database model is introduced to represent contact form submissions.

Each message submitted through the website is stored as a row in the `contact_messages` table.

Typical attributes include:

```
id
name
email
message
created_at
```

This structure ensures that every inquiry submitted through the portfolio can be recorded and reviewed later.

The timestamp field allows messages to be ordered chronologically and supports future administrative features.

---

## Input Validation

Basic validation rules are applied to contact form submissions before they are stored in the database.

Validation checks ensure that submitted messages meet minimum quality and formatting requirements.

Typical validation rules include:

* required field validation
* email format validation
* message length limits

These checks prevent invalid or incomplete submissions from being stored in the database.

Validation occurs at the route level before the message object is created and inserted into the database.

---

## Automatic Timestamping

Each stored contact message automatically records the time at which it was submitted.

The timestamp is generated by the backend at the moment the message is saved to the database.

This allows the application to:

* track when inquiries were received
* sort messages chronologically
* support future administrative interfaces

Automatic timestamping ensures that every submission is recorded with accurate temporal information.

---

## Database Insertion

Once the message passes validation, a new contact message object is created using the SQLAlchemy model.

The message is then inserted into the database using the SQLAlchemy session.

Typical insertion workflow:

```
db.session.add(contact_message)
db.session.commit()
```

After the transaction is committed, the message becomes a persistent record stored in the `contact_messages` table.

---

## Future Administrative Interface

Although not implemented in Phase 3, the database schema is intentionally designed to support a future administrative interface.

In later phases, an admin dashboard may allow the portfolio owner to:

* view contact message submissions
* sort or filter messages
* mark messages as read
* archive or delete entries

Because contact messages are already stored in a structured database table, these features can be added later without modifying the database schema.
