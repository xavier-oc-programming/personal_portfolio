# Phase 3 — Database Integration

## Objective

Replace the in-memory project dataset with a persistent data layer while keeping the same schema and UI behavior.

---

## What this phase consists of

## 1. Database decision and initialization

### 1.1 SQLite database file strategy

Select SQLite as the database engine for the portfolio application.

SQLite provides a lightweight, file-based relational database that is well suited for small web applications and development environments.

Key considerations:

* The database will exist as a **single `.db` file** inside the project.
* No separate database server is required.
* The file can be easily backed up or migrated later if needed.
* The schema must match the **unified project model defined in Phase 1**.

---

### 1.2 Decide whether to use raw SQLite or SQLAlchemy ORM

Determine how the application will interact with the database.

Two options exist:

**Raw SQLite (`sqlite3`)**

* Direct SQL queries
* Simpler for small scripts
* Less abstraction

**SQLAlchemy ORM**

* Object-relational mapping
* Python classes represent database tables
* Cleaner integration with Flask applications
* Easier schema evolution and maintainability

For this project, the selected approach should prioritize:

* maintainability
* scalability
* clean separation between data models and application logic

---

### 1.3 Create schema to match the unified project model

Define the database schema so that it matches the **project structure defined in Phase 1**.

At minimum, the `projects` table should contain fields such as:

* id
* slug
* title
* description
* primary_category
* tags
* tech_stack
* github_url
* live_url
* image_path
* created_at

The goal is to ensure the **database structure aligns exactly with the data model already used by the application**.

This allows the UI layer to remain unchanged while only the data source changes.

---

### 1.4 Decide how to store lists (tags / tech_stack)

Some project attributes are lists rather than single values.

Examples include:

* tags
* tech_stack

Possible storage strategies include:

**Option 1 — JSON string**

Store lists as serialized JSON.

Example:

```
["flask", "bootstrap", "jinja"]
```

Advantages:

* Simple implementation
* Flexible structure

Disadvantages:

* Harder to query directly with SQL

---

**Option 2 — Join tables**

Create relational tables such as:

```
projects
tags
project_tags
```

Advantages:

* Fully normalized schema
* Efficient filtering

Disadvantages:

* More complex implementation

---

**Option 3 — Separate tables**

Store tags or tech stack entries in separate linked tables.

This allows relational filtering while keeping the schema organized.

The chosen strategy should balance **simplicity and future extensibility**.

---

## 2. Project data persistence

### 2.1 Insert projects into the database

Populate the database with the existing portfolio projects.

Possible approaches:

* Create a **seed script**
* Manually insert rows
* Import from a structured dataset

The inserted data must follow the **schema defined earlier**.

---

### 2.2 Replace in-memory list with database queries

Currently, project data is stored in memory using a Python list or dictionary.

This phase replaces that structure with **database queries**.

Instead of:

```
projects = [...]
```

The application will query the database:

```
SELECT * FROM projects
```

Flask routes will retrieve project data dynamically from the database.

---

### 2.3 Ensure sorting and filtering still work

The portfolio site already supports functionality such as:

* category filtering
* tag filtering
* project sorting

These features must continue to work after switching to database queries.

This means:

* converting filtering logic into SQL queries
* ensuring routes return correctly filtered datasets
* validating that the UI behavior remains unchanged

---

## 3. Contact message persistence

### 3.1 Contact form writes to database

The contact form will now persist messages in the database.

A new table should be introduced:

```
contact_messages
```

Typical fields may include:

* id
* name
* email
* message
* created_at

Whenever a user submits the contact form, the message will be inserted into the database.

---

### 3.2 Validation and timestamping

Contact messages must include basic validation.

Examples:

* email format validation
* required fields
* message length limits

Additionally, every message should automatically store a **timestamp** indicating when it was submitted.

---

### 3.3 Optional admin view later

In a future phase, an administrative interface may be added to:

* view contact messages
* manage submissions
* archive or delete entries

This functionality is **not required for Phase 3**, but the database structure should allow for it later.

---

## 4. Phase 3 outputs

### 4.1 Project listings come from DB

The `/projects` page retrieves project data from the database instead of in-memory objects.

---

### 4.2 Detail pages come from DB

Each project detail page loads its content from the database using the project slug or ID.

---

### 4.3 Contact messages persist

Contact form submissions are saved to the `contact_messages` table.

---

### 4.4 Schema remains identical to Phase 1 spec

The database schema mirrors the **unified project model defined during Phase 1**.

This ensures architectural consistency across the application.
