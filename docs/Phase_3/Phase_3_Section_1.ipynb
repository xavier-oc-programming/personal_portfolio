# Phase 3 · Section 1 — Database Foundation

### Objective

Phase 3 Section 1 introduces the database layer for the portfolio application and prepares the system for persistent data storage.

Until this phase, project data was stored in an in-memory Python dataset located in:

```
data/projects.py
```

This section establishes the database architecture, integrates an ORM layer, defines the project schema, and prepares the application for persistent project records.

By the end of this section, the application includes a configured SQLite database, a defined ORM model, and automatic table creation during application startup.

---

## Prerequisites

Before this section, the application already included:

* A working Flask application
* Structured Jinja templates
* A dynamic projects hub and project detail pages
* A unified project dataset in `data/projects.py`

These components were implemented during **Phase 2**.

---

## Implementation Steps

This section introduces the database layer through four steps:

1. Select and configure the database engine.
2. Integrate SQLAlchemy as the ORM layer.
3. Define the `Project` database model.
4. Establish a strategy for storing list-based attributes.

---

## SQLite Database Strategy

The portfolio uses **SQLite** as its database engine.

SQLite is a lightweight, file-based relational database that requires no external server and works well for small applications and development environments.

The database file is stored at:

```
app/database/portfolio.db
```

The application ensures the database directory exists and automatically initializes the database during startup.

---

## ORM Integration

The project integrates **Flask-SQLAlchemy** as the Object-Relational Mapping (ORM) layer.

An ORM allows Python classes to represent database tables, enabling developers to work with database records as Python objects instead of writing raw SQL.

Key benefits:

* cleaner architecture
* separation between routes and database logic
* maintainable model definitions
* easier schema management

Dependency added:

```
Flask-SQLAlchemy
```

---

## Project Database Schema

A new ORM model named **`Project`** represents portfolio entries in the database.

The `projects` table includes fields such as:

* id
* slug
* title
* primary_category
* short_description
* full_description
* featured
* date
* problem
* solution
* challenges
* results
* tags
* tech_stack
* repo_url
* live_url
* demo_url
* card_image
* screenshots
* videos

Database tables are created automatically when the application starts:

```
db.create_all()
```

---

## List Attribute Storage

Some project attributes are naturally lists, including:

* tags
* tech_stack
* screenshots
* videos

To keep the schema simple, these attributes are stored as **JSON strings in text columns**.

Example:

```
["flask", "bootstrap", "jinja"]
```

Helper properties in the ORM model convert these values into Python lists when used inside the application.

## Application Behavior in This Phase

```
data/projects.py
```

This staged approach allows the database infrastructure to be validated before switching the application to database queries.
