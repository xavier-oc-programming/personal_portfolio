## 5 — Upload Result Visuals for Data Projects

**Nobel Prize Analysis and House Price Prediction specifically.**

### Why this matters
Data projects require a different kind of proof than web applications. A recruiter hiring for a data or ML role does not just want to see that you wrote pandas code — they want to see that you derived something meaningful from data and can present it clearly.

A chart uploaded as a screenshot is the fastest way to communicate analytical output to a non-technical reviewer.

### How to export charts from Jupyter

**Option A — Screenshot the rendered output**
1. Run all cells in the notebook
2. Use `Cmd + Shift + 4 → Space` to capture each chart cell output
3. Save as PNG

**Option B — Export via matplotlib**
Add this line after your plot code:
```python
plt.savefig('chart_name.png', dpi=150, bbox_inches='tight')
```

**Option C — Export via Plotly**
```python
fig.write_image('chart_name.png')  # requires kaleido: pip install kaleido
```

- Top 10 countries by total prizes (bar chart)
- Prize count by gender over time (line or bar chart)
