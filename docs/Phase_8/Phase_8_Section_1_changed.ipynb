# Session 2 — March 23 (Partially Uncommitted)

## 7. Local Backup System

A full backup and restore system was built and added to the admin panel.

```bash
app/admin/routes.py                   # 119 lines added
app/templates/admin/backups.html      # New template: backup list with per-row Restore button
app/templates/admin/base.html         # "Backups" link added to sidebar nav
```

### Functions added to routes.py

* `_backup_dir()` — returns path to `data/backup/`
* `_backup_projects()` — writes a timestamped JSON snapshot of all projects to `data/backup/projects_YYYY-MM-DDTHH-MM-SS.json`. Automatically deletes oldest files, keeping only the 3 most recent.

### Routes added

* `GET /admin/backups` — lists all backups with filename, timestamp, and file size
* `POST /admin/backups/restore/<filename>` — restores a backup into the database (upserts by slug). Takes a fresh snapshot of current state before overwriting as a safety measure.
