# Role Positioning

## Purpose

Before writing any copy, the portfolio requires a clear positioning strategy — a defined answer to the question:

> *Who is this person professionally, and what kind of work are they looking for?*

Without this foundation, copy tends to be generic, scattered, and unlikely to resonate with any specific audience.

---

## Target Profile

This portfolio is positioned across a hybrid technical profile combining:

* **Python development** — scripting, automation, backend logic, data pipelines
* **Full-stack development** — web applications, APIs, frontend integration
* **Data analysis** — working with data, deriving insights, presenting results

This is a legitimate and marketable combination. The goal is not to pick one arbitrarily, but to present the combination as a coherent and intentional skill set — the profile of a developer who can build systems *and* work with data end-to-end.

---

## Positioning Statement (Internal Reference)

This is not copy for the site — it is an internal anchor that informs all written content:

```text
A developer who builds full-stack web applications and data tools in Python,
with the ability to handle the full cycle from backend logic to data analysis
and clear output presentation.
```

All copy written in this section should reflect this identity.

# 1.1 — Strong Headline and Role Positioning

## Purpose

The headline is the first thing a recruiter or visitor reads. It must immediately communicate:

* what the person does
* at what level
* across which domains

A weak headline lists a job title. A strong headline communicates professional value.

---

## Headline Strategy

The headline should be:

* specific enough to signal real expertise
* broad enough to reflect the hybrid profile
* human and confident in tone — not a keyword list

---

## Examples

Weak headline:

```text
Python Developer | Data Analyst | Full Stack
```

Strong headline:

```text
Developer building full-stack applications and data tools in Python
```

Or alternatively:

```text
Python developer. I build web apps, automate workflows, and turn data into something useful.
```

The final choice should reflect personal voice while maintaining professional clarity.

---

## Implementation

The headline appears in the following locations:

```bash
app/templates/index.html        # Hero section — primary headline
app/templates/about.html        # Page title or opening statement
```

The hero section should contain:

* a primary headline (one sentence, high impact)
* a supporting subline (one or two sentences expanding on the headline)
* a clear call to action (link to projects or contact)

Example structure:

```html
<h1>Developer building full-stack apps and data tools in Python</h1>
<p>I work across the full stack — from backend APIs to data pipelines — and care about building things that are useful, readable, and well-structured.</p>
<a href="/projects">See my work</a>
```

# 1.2 — Project Descriptions That Emphasize Outcomes and Skills

## Purpose

Project descriptions are typically the first substantive content a recruiter reads after the headline. Most portfolios list what a project *does* — what it *is*. Strong project descriptions communicate:

* what problem was solved
* what was built to solve it
* what technologies and skills it demonstrates
* what the outcome or result was

---

## Description Framework

Each project description should follow this structure:

```text
[Context / Problem] → [What was built] → [Skills demonstrated] → [Outcome or result]
```

This does not need to be four separate sentences — it should flow naturally — but all four elements should be present.

---

## Examples

Weak project description:

```text
A Flask web application with user authentication and a PostgreSQL database.
```

Strong project description:

```text
Built a full-stack Flask application to manage and track personal data across sessions.
Implemented user authentication, a relational database schema, and a clean REST API.
Demonstrates backend architecture, database design, and end-to-end application development.
```

---

## Implementation

Project descriptions are managed in one of two locations depending on implementation:

```bash
app/data/projects.json          # If projects are data-driven
app/templates/project_detail.html  # If content is template-based
```

Each project entry should include:

* a one-line summary (used in cards)
* a full description (used in the detail page)
* an explicit list of skills or technologies demonstrated

Example project data structure:

```json
{
  "title": "Portfolio Site",
  "summary": "A full-stack Flask portfolio built for deployment, measurement, and professional positioning.",
  "description": "Built a production-ready portfolio application using Flask, PostgreSQL, and Jinja2 templates. Deployed to a live server with a custom domain, HTTPS, analytics, and SEO configuration. Demonstrates full-stack development, deployment workflow, and attention to production-level detail.",
  "skills": ["Python", "Flask", "PostgreSQL", "HTML/CSS", "Deployment", "SEO"]
}
```

---

## Audit Checklist

For each project, verify:

* Does the description mention the problem or motivation?
* Does it describe what was built, not just what it does?
* Does it name the skills and technologies in context?
* Does it state or imply a result or outcome?

# 1.3 — About Page That Tells a Coherent Story

## Purpose

The About page is where the professional narrative lives. It is the space that connects the projects, the skills, and the person — and makes the portfolio feel human and intentional rather than mechanical.

A strong About page answers:

* Who is this person?
* How did they get here?
* What do they want to do next?
* Why should someone work with them?

---

## Narrative Structure

The About page should follow a three-part arc:

```text
1. Where you are now — your current identity and what you can do
2. How you got here — the path, interests, and development that brought you here
3. Where you are going — the kind of work or role you are looking for
```

This arc gives the page forward momentum and makes the professional intent explicit without being transactional.

---

## Tone Guidelines

The About page should be:

* written in first person, active voice
* confident but not formal — professional but human
* specific enough to be credible (avoid generic phrases like "passionate about technology")
* honest about the hybrid profile — do not pretend to be a specialist if you are a generalist

---

## Example Structure

```text
I'm a developer with a background across Python, web development, and data analysis.

I got into programming through [origin point — what sparked the interest]. Since then
I've built projects that span backend APIs, full-stack applications, and data pipelines —
because I'm drawn to the end-to-end picture: understanding a problem, building something
to solve it, and seeing the result in the data.

Right now I'm looking for roles where I can [target role context] — whether that's
backend development, data work, or something that sits across both. I care about
writing clean code, building things that actually work, and working with people who
take their craft seriously.
```

---

## Implementation

The About page content lives in:

```bash
app/templates/about.html
```

If the About page uses a data file or context injection:

```bash
app/data/about.json
app/app.py                      # Context passed to template
```

The page should include:

* a primary heading (your name or positioning statement)
* the narrative text in two to four short paragraphs
* a skills or technology section (optional but recommended)
* a link to projects or a contact prompt

# Copy Audit

## Purpose

After updating individual sections, a final audit ensures the portfolio reads consistently as a whole.

---

## Audit Scope

Check the following across all templates:

```bash
app/templates/index.html
app/templates/about.html
app/templates/projects.html
app/templates/project_detail.html
app/templates/contact.html
```

---

## Audit Checklist

**Headline and hero**

* Does the headline immediately communicate what you do?
* Is the subline specific and human — not a list of keywords?
* Is there a clear call to action?

**Project descriptions**

* Does each description follow the outcome-focused framework?
* Are skills named in context rather than in a disconnected list?
* Is the language active and specific?

**About page**

* Does it tell a story, not just list facts?
* Is the target role or direction made clear?
* Is the tone consistent — professional, confident, human?

**Global consistency**

* Does the same professional identity come through on every page?
* Are there contradictions between how the headline and the About page describe you?
* Is the tone consistent across all pages?

---

# Validation

After completing all copy updates, perform the following checks.

## Recruiter Test

Read the portfolio as if you are a recruiter seeing it for the first time:

* Can you understand what this person does within ten seconds?
* Would you know whether they are a fit for a Python developer role?
* Would you know whether they are a fit for a data analyst role?
* Does the About page make you want to reach out?

## Content Check

* Confirm the headline is visible and prominent on the home page
* Confirm each project has a complete, outcome-focused description
* Confirm the About page follows the three-part narrative arc
* Confirm there is no copy that reads as a feature list rather than a value statement
