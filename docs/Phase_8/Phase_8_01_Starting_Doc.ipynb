# Phase 8 — Professional Positioning

## Objective

Make the portfolio **communicate value, not just features**.

The goal of this phase is to transition the portfolio from a technically complete application into a **professionally positioned personal brand** — one that clearly communicates who you are, what you can do, and why you are a strong fit for your target roles.

---

# What this phase consists of

## 1. Copywriting and Narrative

### 1.1 Strong headline and role positioning

The portfolio must open with a headline that immediately communicates professional identity.

A strong headline:

* specifies what you build and with what
* reflects your hybrid profile honestly
* reads as a person, not a keyword list

Example:

```
Developer building full-stack applications and data tools in Python
```

This is the first thing a recruiter reads. It must land in under five seconds.

---

### 1.2 Project descriptions emphasize outcomes and skills

Each project description must communicate more than what the project is — it must communicate what it demonstrates.

Every description should follow this framework:

```
[Context / Problem] → [What was built] → [Skills demonstrated] → [Outcome or result]
```

This shifts the portfolio from a feature archive to a **proof of capability**.

---

### 1.3 About page tells a coherent story for your target role

The About page must function as a professional narrative, not a list of facts.

It should follow a three-part arc:

* **Where you are now** — current identity and what you can do
* **How you got here** — the path and experience that brought you here
* **Where you are going** — the kind of work or role you are looking for

This gives the page forward momentum and makes professional intent explicit.

## 1v2. Section 1 Extensions — Built Beyond Original Spec

The following additions were built during Phase 8 Section 1 and go beyond the original editorial scope. They are documented here as a permanent extension of what Section 1 delivered.

---

### 1v2.1 Full Admin Panel

A complete password-protected admin UI was built from scratch to support ongoing content management of the portfolio without touching the codebase.

The admin panel provides:

* session-based login and logout
* a dashboard with live project, featured, and message counts
* full project CRUD — create, edit, and delete projects through a form UI
* media upload — card images, screenshots, and video files saved to static storage
* contact message viewer with delete capability
* project search by title and filter by category

This transforms the portfolio from a static data file into a **managed content system**.

---

### 1v2.2 Media Upload via Admin UI

The admin panel includes a dedicated media manager for each project.

Supported media types:

* card image — the thumbnail shown on the projects listing page
* screenshots — shown on the project detail page
* video files — for demo recordings

All files are saved to `app/static/images/projects/` and referenced in the database by path.

This enables Phase 8 Section 2 (screenshots and demos) to be completed entirely through the admin UI without any code changes.

---

### 1v2.3 Contact Message Management

The admin panel includes a messages view that displays all contact form submissions.

Capabilities:

* view all messages with sender name, email, subject, and body
* delete individual messages

This closes the loop on the contact form — submissions are now reviewable and manageable from within the application.

---

### 1v2.4 Local Backup System with Restore

A backup and restore system was added to the admin panel to protect project data.

Backup behaviour:

* on demand — creates a timestamped JSON snapshot of all projects in `data/backup/`
* automatically retains only the 3 most recent backups, deleting older ones
* snapshots are human-readable JSON files named `projects_YYYY-MM-DDTHH-MM-SS.json`

Restore behaviour:

* any backup can be restored via the admin UI
* before overwriting, a fresh snapshot of the current state is taken automatically as a safety measure
* restore uses upsert by slug — existing projects are updated, new ones are inserted

This protects against accidental data loss during seeding, editing, or database migrations.

## 2. Proof Signals

### 2.1 Clear tech stack display

Each project must display its technologies clearly and consistently.

Tech stack should be:

* visible on project cards
* specific enough to be scannable by recruiters
* consistent in naming across the site

This allows recruiters to quickly identify relevant skills without reading full descriptions.

---

### 2.2 Clear links to repos

Every project must include a direct link to its GitHub repository.

Requirements:

* repos must be public
* repo names must be readable and professional
* README files should exist for featured projects

This provides immediate access to the actual code — the ultimate proof signal.

---

### 2.3 Screenshots and/or demos

Featured projects must include at least one visual — a screenshot, a demo GIF, or a live link.

Visual proof:

* makes projects feel real and finished
* increases credibility significantly
* gives recruiters something to share internally

A project without a screenshot is invisible to non-technical reviewers.

---

### 2.4 Results and metrics for data projects

Data projects must go beyond showing code — they must show output.

Each data project should include:

* a key finding or insight from the analysis
* a chart, visualisation, or result screenshot
* a one-line summary of what the data revealed

This transforms data work from academic exercises into demonstrated analytical capability.

---

## 3. Phase 8 Outputs

### 3.1 Site reads like a professional profile

By the end of this phase, the portfolio must pass the recruiter test:

* Professional identity is clear within ten seconds
* Target roles are immediately apparent
* Projects communicate skill and impact, not just existence
* About page tells a story rather than listing facts

The site should feel like a **professional profile**, not a project archive.

---

### 3.2 Recruiters can quickly assess fit

The portfolio must be structured so that a recruiter spending sixty seconds on it can answer:

* What does this person build?
* What technologies do they use?
* What is their background?
* Is there a link to their code?
* How do I contact them?

If any of these questions cannot be answered quickly, the portfolio needs further work.
