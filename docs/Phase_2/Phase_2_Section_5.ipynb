# Phase 2 · Section 5 — Project Detail Pages, Category Navigation, and Tag Filtering

### Objective

Phase 2 Section 5 expands the Projects system from a simple project grid into a structured browsing experience.

This section introduces **project detail pages**, **dedicated category routes**, and **tag-based filtering**, allowing visitors to explore projects through stable URLs and flexible filters.

By the end of this section, the portfolio supports category navigation, tag exploration, slug-based project pages, and consistent filtering and sorting across all views.

---

## Category Routes

Dedicated category routes are introduced for the three portfolio pillars:

```
/projects/web
/projects/data
/projects/software
```

Each route renders the same `projects.html` template while filtering the dataset by `primary_category`.

This improves navigation, URL clarity, and maintainability without creating duplicate templates.

---

## Project Detail Pages

A slug-based route allows each project to have its own detail page:

```
/projects/<slug>
```

Examples:

```
/projects/portfolio-website
/projects/morse-code-converter
```

The `project_detail.html` template displays full project information including:

* title and description
* category and tag badges
* problem, solution, challenges, and results
* tech stack
* GitHub repository link
* screenshots and video sections

Invalid slugs trigger the application's custom `404.html` page.

---

## Tag Filtering

Tag-based filtering is introduced to allow projects to be browsed by technology or theme.

Examples:

```
/projects?tag=flask
/projects/web?tag=flask
/projects/software?tag=python
```

Tag filtering works alongside category filtering and sorting, enabling combined queries such as:

```
/projects/web?tag=flask&sort=newest
```

Available tags are dynamically derived from the project dataset.

---

## Projects Hub Improvements

The `projects.html` template is expanded to support:

* category navigation buttons
* dynamic tag filter controls
* sorting controls
* active-state styling for filters
* reusable project cards with clickable tag badges

Page headings and descriptive text adapt automatically based on the active category or tag.

---

## Unified Filtering and Sorting Logic

Filtering and sorting logic is centralized in helper functions inside `app.py`.

A shared context builder prepares the data used by all project views, including:

```
projects
active_category
active_tag
active_sort
category_counts
available_tags
```

This keeps route logic clean and ensures consistent behavior across category pages and the main Projects hub.

---

## Files Updated

Python logic:

```
app/app.py
```

Templates:

```
templates/projects.html
templates/project_detail.html
templates/components/project_card.html
```

Template files were also standardized with top-level documentation comments.

---

## Result

With Section 5 complete, the portfolio now supports a structured project browsing system with category navigation, tag filtering, project detail pages, and robust URL-based filtering.

These improvements significantly enhance usability while preserving the reusable template architecture introduced earlier in Phase 2.

This prepares the project for **Phase 3**, where the in-memory project dataset will be replaced by a persistent database layer.
