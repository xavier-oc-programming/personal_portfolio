# Phase 2 · Section 3 — MVP UI Styling

### Objective

Phase 2 Section 3 introduces the **Minimum Viable Product (MVP) user interface styling layer** for the portfolio application.

This section establishes a consistent visual and layout baseline using **Bootstrap and a small set of custom CSS rules**. The goal is not to fully design the site yet, but to provide a stable structure that ensures pages are readable, consistent, and responsive across screen sizes.

By the end of this section, all page templates share a unified layout system with consistent typography, spacing, and responsive behavior, preparing the application for dynamic project rendering in later stages.

---

## Bootstrap Layout Integration

The application integrates **Bootstrap via CDN** to quickly establish a responsive layout system.

Bootstrap provides a grid-based structure using containers, rows, and columns, allowing pages to adapt naturally to different screen sizes.

The global layout now uses a container-based structure to constrain page width and maintain visual balance.

This approach provides several benefits:

* consistent horizontal alignment
* predictable responsive behavior
* faster UI development during early project stages

Bootstrap is loaded globally in `base.html`, ensuring that all templates automatically inherit the same layout system.

---

## Typography and Layout Baseline

A basic typography hierarchy is introduced to ensure consistency across the application.

Pages follow a simple structure:

* **H1** for the page title
* **H2 / H3** for section headings
* standard paragraph text for body content

Spacing between sections is standardized using a reusable layout helper defined in the global stylesheet:

```
.page-section {
    margin-bottom: 3rem;
}
```

Templates use this class to maintain a consistent vertical rhythm between content sections.

This approach reduces repeated spacing utilities and centralizes layout behavior within the stylesheet.

---

## Responsive Template Structure

All page templates are updated to follow a consistent section layout using Bootstrap spacing utilities and semantic sections.

Each page now follows a predictable pattern:

```
<section class="page-section">
    <h1>Page Title</h1>
</section>

<section class="page-section">
    <h2>Section Title</h2>
</section>
```

This ensures that content remains readable and visually balanced across devices while keeping templates simple and maintainable.

---

## Static Assets Pipeline

The project introduces a structured **static assets directory** for CSS, images, and optional JavaScript.

The static folder now follows this layout:

```
static/
│
├── css/
│   └── styles.css
│
├── images/
│   ├── profile/
│   │   └── headshot.png
│   │
│   └── projects/
│       └── morse_code.png
│
└── js/
    └── main.js
```

This structure separates styling, images, and scripts into clear categories and ensures that all assets can be referenced reliably using Flask’s `url_for('static', ...)` helper.

Initial image assets are added to support future UI enhancements such as project cards and profile sections.

---

## Template Updates

All page templates are updated to align with the new layout and styling conventions.

Updated templates include:

```
templates/
    index.html
    about.html
    projects.html
    project_detail.html
    contact.html
    404.html
```

Each page now uses consistent spacing, heading hierarchy, and layout structure.

These updates ensure that the entire site maintains a uniform appearance before dynamic project content is introduced.

---

## Result

With Section 3 complete, the portfolio application now has a stable **MVP UI styling foundation**.

The site features a consistent layout system, standardized typography and spacing, responsive behavior, and a structured static assets pipeline.

While the design remains intentionally minimal, the application now provides a clean and professional interface that supports future development.

This foundation prepares the project for the next stage of Phase 2, where project data will be rendered dynamically and the portfolio will evolve into a data-driven application.
