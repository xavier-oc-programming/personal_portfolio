# Phase 2 · Section 1 — Application Foundation

### Objective

Phase 2 Section 1 establishes the foundational structure of the portfolio web application. This section introduces the Flask application entrypoint, centralized configuration management, and the core navigation routes that define the site's basic structure.

By the end of this section, the project transitions from a static repository into a working Flask application capable of serving pages locally in a stable and reproducible environment.

---

## Flask Application Initialization

The application is initialized through a dedicated entrypoint located in `app/app.py`.

The project uses the **Application Factory Pattern**, which encapsulates the creation of the Flask application inside a function. This pattern improves modularity and prepares the application for future expansion.

Benefits include:

* clear separation between initialization and route logic
* improved maintainability
* compatibility with future architectural features such as Blueprints

---

## Centralized Configuration

Application settings are managed through a dedicated configuration module located in `app/config.py`.

The configuration system supports multiple environments through separate configuration classes:

* **Development configuration** — enables debugging and auto-reload
* **Production configuration** — disables debugging for deployment

Important configuration values include:

* `DEBUG`
* `SECRET_KEY`

Configuration values are loaded through environment variables to avoid hard-coding sensitive information in the source code.

---

## Environment Variable Integration

Environment variables are managed using a `.env` file during development.

Typical variables include:

* `FLASK_ENV`
* `FLASK_DEBUG`
* `SECRET_KEY`

The `.env` file is loaded at application startup, allowing the configuration module to read these values and configure the application accordingly.

This keeps sensitive configuration outside the codebase while maintaining a simple local development workflow.

---

## Core Route Skeleton

The application defines the core navigation routes that represent the main sections of the portfolio site.

These routes include:

* `/` — Home page
* `/about` — About page
* `/projects` — Portfolio projects hub
* `/projects/<slug>` — Dynamic project detail page
* `/contact` — Contact page

Each route renders a template from the `templates` directory, establishing the initial connection between the Flask backend and the front-end template system.

---

## Error Handling

A custom **404 error handler** is implemented to manage invalid routes.

When a user accesses a non-existent page, the application renders a dedicated 404 template instead of returning a generic server response.

This improves the user experience and ensures proper HTTP status handling.

---

## Local Development Reliability

The application can be started locally using:

```
python app/app.py
```

Verification confirms that:

* the server starts successfully
* core routes load without errors
* templates render correctly
* static assets are served properly
* debug mode automatically reloads the server when files change

---

## Result

With Section 1 complete, the portfolio project now has a functioning Flask backend with a stable initialization process, centralized configuration, and a complete route skeleton.

This foundation provides the base architecture required for implementing dynamic project content, database integration, and additional application features in later phases.
