# Phase 2 · Section 2 — Jinja Template System

### Objective

Phase 2 Section 2 introduces a structured template architecture for the portfolio application using **Flask's Jinja templating system**.

This section establishes a reusable layout system that separates shared interface elements from page-specific content. The result is a maintainable template structure where common UI components are defined once and reused across the entire site.

By the end of this section, the application transitions from isolated page templates to a unified layout system powered by template inheritance.

---

## Global Layout Template

A global layout template is implemented in `templates/base.html`.

This template defines the universal structure of every page on the site, including:

* the HTML document structure
* global meta tags
* stylesheet loading
* navigation and footer placement
* the primary content injection point

The template exposes a Jinja content block:

```
{% block content %}{% endblock %}
```

All page templates inherit from this base layout, ensuring a consistent page structure throughout the application.

---

## Template Inheritance

Each page template now extends the base layout using Jinja’s template inheritance mechanism:

```
{% extends "base.html" %}
```

Page-specific content is inserted into the layout through defined blocks such as:

```
{% block title %}
{% block content %}
```

This approach prevents duplication of shared markup and allows the layout to evolve without requiring changes to every page.

---

## Reusable UI Components

Shared interface elements are extracted into reusable template components located in:

```
templates/components/
```

These include:

* `navbar.html`
* `footer.html`
* `project_card.html` (placeholder for later use)

The base layout incorporates these components using:

```
{% include "components/navbar.html" %}
{% include "components/footer.html" %}
```

This modular structure ensures that updates to navigation or footer elements automatically propagate across all pages.

---

## Page Template Structure

Dedicated templates are implemented for each major section of the site:

```
templates/
    index.html
    about.html
    projects.html
    project_detail.html
    contact.html
    404.html
```

Each template defines its own title and content while inheriting the global layout.

Page content is organized into semantic sections using headings and spacing utilities, preparing the site for future styling improvements.

## Template Organization

With this section complete, the template system now follows a clear hierarchy:

```
templates/
│
├── base.html
├── index.html
├── about.html
├── projects.html
├── project_detail.html
├── contact.html
├── 404.html
│
└── components/
    ├── navbar.html
    ├── footer.html
    └── project_card.html
```

This architecture separates layout, reusable UI elements, and page-specific templates.

With Section 2 complete, the portfolio application now uses a scalable Jinja template architecture based on inheritance and reusable components.
