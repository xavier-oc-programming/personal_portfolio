# 0. Basic Setup & Environment Foundation

Phase 2 begins by establishing a controlled and reproducible development environment.
No application logic is written before the environment, version control structure, and tooling discipline are formally defined.

This section ensures that all future development occurs within a stable, traceable framework.

---

## 0.1 Repository Structure Initialization

Before writing any code, the repository structure must be intentional and clean.

The root of the project is organized as follows:

```
PROJECT_ROOT/
│
├── .gitignore
├── .env
├── .python-version
├── requirements.txt
├── README.md
│
├── app/
├── docs/
├── Phase_1/
├── Phase_2/
```

The `app/` directory contains runtime application logic.
Documentation and planning artifacts are kept separate from executable code.

This separation enforces architectural clarity from the outset.

---

## 0.2 Virtual Environment Creation

All development occurs inside an isolated Python virtual environment.

The environment is created at the project root:

```bash
python3 -m venv .venv
```

Activated with:

```bash
source .venv/bin/activate
```

The `.venv/` directory:

* Exists locally
* Is ignored by Git
* Ensures dependency isolation
* Prevents system-level contamination

This guarantees that the project’s runtime behavior is independent of the host machine’s global Python state.

---

## 0.3 Dependency Installation

Phase 2 intentionally limits dependencies to only what is required for the MVP core website.

Installed packages:

```bash
pip install Flask python-dotenv
```

No database, form, authentication, or deployment libraries are introduced at this stage.

This enforces minimal surface area and reduces architectural complexity during the MVP phase.

---

## 0.4 Dependency Snapshot (Reproducibility)

After installation, the exact dependency tree is frozen:

```bash
pip freeze > requirements.txt
```

This file captures:

* Direct dependencies (Flask, python-dotenv)
* Transitive dependencies (Jinja2, Werkzeug, etc.)
* Exact version numbers

Reconstruction on another machine requires only:

```bash
pip install -r requirements.txt
```

This establishes reproducibility as a first-class concern.

---

## 0.5 Environment Variable Configuration

Configuration is separated from application code via a local `.env` file:

```
FLASK_ENV=development
SECRET_KEY=dev-secret-key
```

The `.env` file:

* Is ignored by Git
* Contains no sensitive production credentials
* Establishes a scalable configuration pattern for later phases

This separation enforces clean configuration management.

---

## 0.6 Python Version Declaration

The interpreter version used for development is declared in:

```
.python-version
```

The file contains only:

```
3.12.5
```

This ensures:

* Interpreter consistency
* Tool compatibility (pyenv, VS Code)
* Protection against version mismatch issues

Unlike `.env`, this file is committed to the repository.

---

## 0.7 Local Environment Verification

Before initializing version control, the environment must be validated:

* The virtual environment activates correctly.
* `python --version` matches `.python-version`.
* `pip freeze` reflects only expected packages.
* The root directory is clean and intentional.

Only after the environment is verified do we proceed to version control initialization.

---

## 0.8 Git Initialization (Local)

Git is initialized at the project root:

```bash
git init
```

This establishes local version control tracking.

---

## 0.9 Initial Staging Verification

Files are staged intentionally:

```bash
git add .
```

Verification via:

```bash
git status
```

Confirms:

* `.venv` and `.env` are ignored
* `.python-version` is staged
* `requirements.txt` is staged
* `.gitignore` and `README.md` are staged

If incorrect files appear, `.gitignore` is corrected before proceeding.

---

## 0.10 Baseline Commit

The initial commit documents the architectural baseline:

```bash
git commit -m "Initial project setup (Phase 2 section 0): environment foundation and base structure"
```

This commit represents a stable, reproducible starting point before application logic begins.

An empty repository is created on GitHub and linked:

From this point forward:

* `main` must always remain stable
* Development occurs in feature or phase branches

## 0.12 Branching Strategy for Phase Development

Each major phase is developed in its own branch.

For Phase 2:

```bash
git checkout -b phase-2-core-mvp
```

This isolates work and preserves historical clarity.

## 0.13 Commit Discipline

Commits are created at logical architectural milestones, not arbitrarily.

Examples:

* Add Flask application skeleton

### 1. Application foundation

#### 1.1 Flask app initialization and config setup

* Create the Flask app entrypoint (`app.py` or `run.py`) with a minimal, stable structure.
* Establish configuration patterns early:

  * `DEBUG` toggle (local dev)
  * `SECRET_KEY` placeholder (needed later for forms/sessions)
  * Clear separation between app init and route logic (even if still small).
* Confirm the “single source of truth” for starting the app:

  * `python app.py` (or equivalent), consistent every time.

#### 1.2 Route skeleton for core pages

* Implement the core navigation routes as real working endpoints:

  * `/` → Home
  * `/about` → About
  * `/projects` → Projects hub
  * `/contact` → Contact
* Each route renders a real template file (no inline HTML).
* Each route returns HTTP 200 and a page with shared layout.

#### 1.3 Environment setup and dependency management

* Create and use a local virtual environment for this project.
* Install Flask (and Bootstrap only if you decide to use it).
* Maintain `requirements.txt` so the project can be cloned and run cleanly.
* Keep environment steps reproducible:

  * create venv
  * install requirements
  * run app

#### 1.4 Confirm server runs locally reliably

* Confirm:

  * No template errors
  * No missing static assets
  * No route breaks
  * Reload works in debug mode
* Establish “known-good startup state”:

  * server starts within a few seconds
  * `/`, `/about`, `/projects`, `/contact` load without errors

---

### 2. Template system (Jinja foundation)

#### 2.1 `base.html` with global layout, navbar, footer

* Build a single global layout template that all pages inherit from.
* `base.html` must include:

  * `<head>` with title block, meta tags
  * CSS links (Bootstrap CDN or local CSS)
  * Header navbar include
  * Main content block
  * Footer include
* This is the “layout contract” all pages must follow.

#### 2.2 Template inheritance across all pages

* Every page template extends `base.html`.
* Each page fills:

  * page title block (optional but recommended)
  * main content block (required)
* No page should rebuild header/footer manually.

#### 2.3 Shared components: nav, footer, project card partial (optional)

* Create template partials in `templates/components/`:

  * `navbar.html`
  * `footer.html`
  * optional: `project_card.html` for consistent project preview rendering
* Components reduce duplication and force consistency.

#### 2.4 Clear page-level templates: index/about/projects/contact

* Create explicit templates:

  * `home.html`
  * `about.html`
  * `projects.html`
  * `contact.html`
* Each page has real content structure (headings, sections, spacing).
* No “blank placeholder” pages in the MVP.

---

### 3. MVP UI styling

#### 3.1 Bootstrap layout grid

* Decide your baseline layout system:

  * Bootstrap CDN (fastest MVP)
  * or custom CSS with container/grid (slower, but fine if you want control)
* If Bootstrap:

  * use container + rows + columns consistently
  * keep spacing utilities consistent (margin/padding scale)

#### 3.2 Typography and spacing baseline

* Define typography conventions:

  * page titles (H1)
  * section headers (H2/H3)
  * body text sizing
  * link style
* Define spacing conventions:

  * vertical rhythm (consistent spacing between sections)
  * card padding
  * layout padding (top/bottom)

#### 3.3 Simple responsive behavior

* Navbar behaves well on smaller widths (Bootstrap default is fine).
* Project cards wrap properly.
* Text blocks do not overflow and remain readable.

#### 3.4 Static assets pipeline: css folder, images, optional JS

* Create `static/` structure:

  * `static/css/`
  * `static/images/`
  * optional: `static/js/`
* Ensure:

  * CSS loads correctly
  * images can be referenced reliably
  * no broken static paths

---

### 4. Projects hub implementation (key feature of Phase 2)

#### 4.1 Use a temporary in-memory dataset (Python list/dict) that follows your schema

* Projects must live in a structured dataset (not hardcoded in templates).
* Use a Python structure in `data/projects.py` (or similar) containing a list of dicts.
* Each project follows your Phase 1 schema direction:

  * unique `slug`
  * `title`
  * `primary_category` (Web/Data/Software)
  * `tags` (list)
  * `short_description`
  * `full_description`
  * `tech_stack`
  * `links` (repo/live/demo)
  * optional: image path
  * optional: `featured` boolean

#### 4.2 Render project cards via Jinja loop

* `/projects` page renders projects using a loop:

  * one card per project
  * consistent layout
  * clear CTA link to detail page
* Optional: use a component partial (`project_card.html`) to keep cards uniform.

#### 4.3 Sorting: A–Z, newest, oldest

* Sorting behavior defined at the route level, not in templates.
* Sorting modes:

  * A–Z by title
  * newest first (requires date field, or a proxy ordering field)
  * oldest first
* A sorting UI control exists (dropdown or buttons), even if simple.

#### 4.4 Filtering by `primary_category`

* Filtering by Web/Data/Software must exist.
* The category filter can be:

  * querystring (recommended): `/projects?category=web`
  * or dedicated routes (also allowed, see 4.5)
* Filter behavior:

  * only show matching projects
  * fallback to “all projects” when filter is missing/invalid

#### 4.5 Dedicated category pages: web/data/software views

* Provide dedicated category entry points:

  * `/projects/web`
  * `/projects/data`
  * `/projects/software`
* These routes:

  * reuse the same template as Projects hub
  * provide a pre-filtered dataset
* The navigation/UX makes it obvious these are category views.

#### 4.6 Featured projects on Home pulled from `featured=True`

* Home page includes a “Featured Projects” section.
* This section:

  * selects projects where `featured == True`
  * renders them using the same card style
* Featured is a curated subset (not all projects).

---

### 5. Project detail pages

#### 5.1 Route: `/projects/<slug>` or `/project/<slug>`

* Use slug-based routing so every project has a stable URL.
* Route must:

  * look up project by slug
  * return 404 for invalid slug (clean behavior)

#### 5.2 Detail template renders full_description, tech_stack, tags, links, image

* Detail page displays:

  * title
  * full description
  * tech stack list
  * tags
  * links (GitHub, live demo, etc.)
  * image (if provided)
* Layout should feel like a “case study page” even in MVP form.

#### 5.3 Navigation back to Projects hub

* Provide a clear way back:

  * “Back to Projects”
  * optionally also “Back to category”
* Avoid dead ends.
