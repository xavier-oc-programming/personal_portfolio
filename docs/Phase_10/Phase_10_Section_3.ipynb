# 3.3 — Open Graph and Social Sharing

## What Open Graph does

When you share a link on LinkedIn, Twitter, Slack, or WhatsApp, the platform reads the Open Graph tags from the page to generate a preview card — title, description, and image. Without these tags, the preview is blank or generic.

## Meta tags to add to base.html

```html
<head>
  <!-- Standard -->
  <meta name="description" content="{{ meta_description or config.DEFAULT_META_DESCRIPTION }}">

  <!-- Open Graph -->
  <meta property="og:title" content="{{ page_title or config.SITE_NAME }}">
  <meta property="og:description" content="{{ meta_description or config.DEFAULT_META_DESCRIPTION }}">
  <meta property="og:image" content="{{ config.SITE_URL }}/static/images/og-image.png">
  <meta property="og:url" content="{{ canonical_url }}">
  <meta property="og:type" content="website">

  <!-- Twitter Card -->
  <meta name="twitter:card" content="summary_large_image">
  <meta name="twitter:title" content="{{ page_title or config.SITE_NAME }}">
  <meta name="twitter:description" content="{{ meta_description or config.DEFAULT_META_DESCRIPTION }}">
  <meta name="twitter:image" content="{{ config.SITE_URL }}/static/images/og-image.png">
</head>
```

## og:image spec

* Size: **1200 × 630px**
* Format: PNG or JPG
* Content: name, role, site URL — keep it clean and readable at small sizes
* Save to: `app/static/images/og-image.png`

# 3.4 — Sitemap and robots.txt

## sitemap.xml

A sitemap tells search engines which pages exist. Generate it dynamically:

```python
from flask import make_response, render_template

@app.route('/sitemap.xml')
def sitemap():
    projects = Project.query.all()
    response = make_response(render_template('sitemap.xml', projects=projects))
    response.headers['Content-Type'] = 'application/xml'
    return response
```

```xml
<!-- templates/sitemap.xml -->
<?xml version="1.0" encoding="UTF-8"?>
<urlset xmlns="http://www.sitemaps.org/schemas/sitemap/0.9">
  <url><loc>{{ config.SITE_URL }}/</loc></url>
  <url><loc>{{ config.SITE_URL }}/projects</loc></url>
  <url><loc>{{ config.SITE_URL }}/about</loc></url>
  <url><loc>{{ config.SITE_URL }}/contact</loc></url>
  {% for project in projects %}
  <url><loc>{{ config.SITE_URL }}/projects/{{ project.slug }}</loc></url>
  {% endfor %}
</urlset>
```

## robots.txt

```python
@app.route('/robots.txt')
def robots():
    content = """User-agent: *
Allow: /
Disallow: /admin/

Sitemap: {}/sitemap.xml""".format(app.config['SITE_URL'])
    response = make_response(content)
    response.headers['Content-Type'] = 'text/plain'
    return response
```
