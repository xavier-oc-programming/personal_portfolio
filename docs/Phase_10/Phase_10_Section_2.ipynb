# 2.1 — Visual Polish

Establish a consistent typographic scale across all pages:

```css
/* Heading hierarchy */
h1 { font-size: 2.5rem; font-weight: 700; }
h2 { font-size: 1.75rem; font-weight: 600; }
h3 { font-size: 1.25rem; font-weight: 600; }

/* Body text */
body { font-size: 1rem; line-height: 1.7; }

/* Section spacing */
.section { padding: 4rem 0; }
```

```css
:root {
  --color-primary: #your-primary;
  --color-accent: #your-accent;
  --color-text: #212529;
  --color-muted: #6c757d;
  --color-bg: #ffffff;
  --color-surface: #f8f9fa;
}
```

# 2.3 — Loading and Feedback States

## Skeleton loader for projects page

While the API fetches project data, show placeholder cards instead of a blank page:

```css
.skeleton {
  background: linear-gradient(90deg, #f0f0f0 25%, #e0e0e0 50%, #f0f0f0 75%);
  background-size: 200% 100%;
  animation: shimmer 1.5s infinite;
  border-radius: 4px;
}

@keyframes shimmer {
  0% { background-position: 200% 0; }
  100% { background-position: -200% 0; }
}
```

```html
<!-- Shown before projects load -->
<div id="skeleton-container" class="row g-4">
  {% for _ in range(6) %}
  <div class="col-md-4">
    <div class="card h-100">
      <div class="skeleton" style="height: 180px;"></div>
      <div class="card-body">
        <div class="skeleton mb-2" style="height: 1.2rem; width: 70%;"></div>
        <div class="skeleton" style="height: 0.9rem; width: 90%;"></div>
      </div>
    </div>
  </div>
  {% endfor %}
</div>
```

In `projects-filter.js`, hide the skeleton and show real cards once the API responds.

Disable the submit button and show a spinner while the form submits:

# 2.4 — Custom Error Pages

## Templates

```html
<!-- templates/errors/404.html -->
{% extends 'base.html' %}
{% block content %}
<div class="text-center py-5">
  <h1 class="display-1 fw-bold">404</h1>
  <p class="lead">{{ _("This page doesn't exist.") }}</p>
  <a href="{{ url_for('index') }}" class="btn btn-primary">{{ _("Go home") }}</a>
</div>
{% endblock %}
```

```html
<!-- templates/errors/500.html -->
{% extends 'base.html' %}
{% block content %}
<div class="text-center py-5">
  <h1 class="display-1 fw-bold">500</h1>
  <p class="lead">{{ _("Something went wrong on our end.") }}</p>
  <a href="{{ url_for('index') }}" class="btn btn-primary">{{ _("Go home") }}</a>
</div>
{% endblock %}
```

## Register in app.py

```python
@app.errorhandler(404)
def not_found(e):
    return render_template('errors/404.html'), 404

@app.errorhandler(500)
def server_error(e):
    return render_template('errors/500.html'), 500
```
