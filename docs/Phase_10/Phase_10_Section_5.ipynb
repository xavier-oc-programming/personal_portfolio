# Phase 10 · Section 5 — Visibility

### Objective

Get the portfolio in front of the right people. The project is technically strong and visually polished — this section ensures it is seen.

A portfolio that nobody sees is just a private project. Visibility is what converts it into career leverage.

---

## Prerequisites

* Open Graph tags configured — from Section 3
* Portfolio live at public URL
* LinkedIn profile active
* GitHub profile active with portfolio repo public

1. Polish the GitHub repository profile — README, repo description, topics
2. Pin the portfolio repo on your GitHub profile
3. Update your GitHub profile README if you have one

5. Update your LinkedIn profile to link to the live site

Add the live URL so visitors can click directly to the site from the repo.

# 5.3 — LinkedIn Announcement

## Why this matters

A LinkedIn post announcing the portfolio will reach hiring managers, developers, and recruiters who would never find the GitHub repo on their own. It also signals that you are actively building and sharing your work — which is itself a positive signal.

The Open Graph tag you configured in Section 3 means the post will show a proper preview card with image, title, and description.

## Post structure

A strong announcement post follows this structure:

```
Hook (first line — visible before "see more")
  Make it specific, not generic.
  Example: "I just shipped a production-grade portfolio built entirely in Python and Flask."

What you built (2–4 lines)
  Describe the technical highlights concisely.
  Focus on what makes it different from a typical portfolio.

What you learned (1–2 lines)
  One genuine insight from the build.

Call to action
  Link to the live site and/or GitHub repo.
  Invite feedback — posts with replies get more reach.

Hashtags (3–5)
  #Python #Flask #Portfolio #WebDevelopment #OpenToWork (if applicable)
```

## Example draft

```
I just shipped a production-grade portfolio built entirely in Python and Flask.

Not a template. Not a no-code builder. A full-stack web application — the kind
you'd build professionally.

What's under the hood:
→ Flask backend with PostgreSQL
→ Password-protected admin dashboard for managing projects without touching code
→ Public REST API exposing all project data
→ Docker for reproducible local setup
→ GitHub Actions CI/CD — tests run on every push, Railway deploys automatically
→ 26 automated tests covering routes, auth, and API endpoints
→ Bilingual — English and Spanish

The portfolio is the proof of skill.

Live: xavieroc.dev
Repo: github.com/xavier-oc-programming/personal_portfolio

Happy to get feedback from anyone who wants to take a look.

#Python #Flask #Portfolio #WebDevelopment #100DaysOfCode
```
