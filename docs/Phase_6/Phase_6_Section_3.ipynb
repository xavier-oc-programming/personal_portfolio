## Static Asset Caching

Browser caching is introduced to improve performance for returning visitors.

Static files such as CSS, JavaScript, and images are configured to be cached for a defined period.

Implementation:

```python
SEND_FILE_MAX_AGE_DEFAULT = timedelta(days=30)
```

This ensures that:

* assets are not re-downloaded on every visit
* page load times are reduced significantly after the first load
* bandwidth usage is minimized

## Image Optimization

Images are optimized both at the **file level** and **template level**.

### Template-level improvements

Images now include:

```html
loading="lazy"
decoding="async"
```

This ensures that:

* images load only when needed (lazy loading)
* decoding happens asynchronously for smoother rendering

### Asset-level improvements

Image files are standardized to appropriate dimensions:

* card images → ~1200 × 800 px
* detail screenshots → max ~1600 px width

Formats are optimized:

* prefer **WebP** for better compression
* use PNG only when necessary

These improvements ensure:

* faster page load times
* reduced file sizes
* consistent visual presentation

## JavaScript Efficiency

JavaScript is refined to improve performance and avoid unnecessary work during runtime.

Example improvement:

* replacing inline style updates with class-based toggling
* using passive scroll listeners for better performance

Example:

```javascript
btn.classList.toggle("is-visible", window.scrollY > 300);
```

This ensures:

* smoother scrolling behavior
* cleaner separation between CSS and JavaScript
* more maintainable interaction logic

## External Resource Optimization

External resources (Bootstrap CDN) are optimized using connection hints.

Implementation in `base.html`:

```html
<link rel="preconnect" href="https://cdn.jsdelivr.net" crossorigin>
<link rel="dns-prefetch" href="https://cdn.jsdelivr.net">
```

This allows the browser to:

* establish connections earlier
* reduce latency when loading external CSS/JS
