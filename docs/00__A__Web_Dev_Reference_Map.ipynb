## How I Built My Portfolio Website Using the 100 Days of Code Curriculum

This document outlines how I leveraged different sections of the **100 Days of Code (Python)** curriculum to build my Day 83 Portfolio Website.

Rather than treating each lesson as isolated knowledge, I approached this project by asking:

> Where do I pull knowledge from?

This portfolio represents the integration of multiple modules across frontend, backend, data analysis, deployment, and professional tooling.



---

# I. Core Web Foundations

These lessons provided the direct backend structure of the website.



## Days 55–58 — Flask Basics

From these lessons, I implemented:

* Flask app initialization
* Routing with `@app.route`
* Returning HTML responses
* Running a local development server
* Debug configuration

Used for:

* Creating `app.py`
* Defining core routes (home, about, projects, contact)
* Structuring backend logic


---

## Day 59 — Jinja Templates

This module introduced dynamic template rendering.

Key concepts applied:

* `render_template()`
* Passing variables to HTML
* Template inheritance
* `{% block content %}`
* `{{ variable }}` syntax

Used for:

* Creating a reusable `base.html`
* Maintaining separation of logic and presentation
* Dynamically rendering projects



---

## Day 60 — HTML Forms

This lesson enabled backend form handling.

Applied concepts:

* `<form>` elements
* POST requests
* Handling form data in Flask

Used for:

* Contact form implementation
* Processing user submissions

# Development Approach

Rather than treating lessons as isolated units, I structured this project by:

1. Identifying required functionality
2. Mapping each feature to specific curriculum modules
3. Integrating concepts across frontend, backend, and data layers
4. Building incrementally from MVP to advanced features

This project represents the consolidation of everything learned from Days 1–83.
