# Phase 8 · Section 2 — Proof Signals

## Objective

Make the portfolio **show evidence, not just claims**.

Phase 8 Section 1 established a professionally positioned narrative — strong headlines, outcome-focused project descriptions, and a coherent About page story. The portfolio now communicates identity and intent clearly.

Phase 8 Section 2 adds the proof layer.

Proof signals are the concrete, verifiable evidence that what the portfolio claims is real. They answer the recruiter's implicit question: *"Okay, but can I actually see it?"*

This section covers four deliverables:

* Tech stack visibility — clear, consistent display on all project cards and detail pages
* GitHub repo links — direct, public access to the code behind every project
* Screenshots and demos — at least one visual for every featured project
* Results and metrics — key findings and output visuals for data and ML projects

By the end of this section, the portfolio passes the non-technical reviewer test: a recruiter or hiring manager with no engineering background can look at the portfolio, understand what was built, and see evidence that it works.

# 2.1 — Clear Tech Stack Display

## Purpose

Recruiters and hiring managers often scan portfolios looking for specific technology names. If a project uses Flask, PostgreSQL, pandas, or scikit-learn, those names need to be visible without reading a paragraph.

Tech stack display is the most efficient signal in the portfolio. It allows a technical recruiter to assess fit in seconds.

---

## Requirements

* Technologies displayed on project cards — visible in the listing view without clicking through
* Technologies displayed on project detail pages — in a dedicated, scannable block
* Naming is consistent across the site — "Flask" not "flask", "PostgreSQL" not "postgres"
* Stack reflects what was actually used — no inflation or padding

---

## Implementation

Tech stack data is stored in the project entry under the `tags` or `tech_stack` field in the database and populated via the admin panel.

Display locations:

```bash
app/templates/projects.html         # Project card — tag pills below title
app/templates/project_detail.html   # Detail page — tech stack section
```

Each technology should render as a styled tag or pill — visually distinct from body text and scannable at a glance.

---

## Audit Checklist

* Every project card shows at least two technology tags
* Every detail page has a dedicated tech stack block
* Technology names are capitalised consistently across all projects
* No project shows a tag that does not reflect actual use in the project

---

# 2.2 — Clear Links to Repos

## Purpose

A portfolio without code access is a claim without evidence. GitHub links are the direct path to the work itself — the ultimate proof signal for any technical role.

Recruiters at technical companies will check repos. Hiring managers will share repo links internally. The repos need to be ready.

---

## Requirements

### Repository standards

* All featured project repos must be public
* Repo names must be readable and professional — lowercase, hyphenated, descriptive
* Each featured project must have a README that describes what the project is and how to run it

### README minimum content

* Project title and one-line description
* Technologies used
* How to install and run locally
* Screenshot or demo link (if available)
* Brief notes on project structure or design decisions

### Link placement on the portfolio

* GitHub link visible on the project card — icon or label
* GitHub link prominent on the project detail page — above the fold
* Live demo link shown where applicable

---

## Implementation

Repo links are stored in the project entry under the `github_url` field and managed through the admin panel. No code changes required.

Verify each link resolves to a public repository before marking the project complete.

---

## Audit Checklist

* All featured projects have a public GitHub repo
* All featured project repos have a README
* READMEs include title, tech stack, and setup instructions
* GitHub links are visible and functional on both card and detail views
* Repo names are professional and readable

# 2.3 — Screenshots and Demos

## Purpose

A project without a screenshot is invisible to non-technical reviewers. Visual proof is what makes a project feel real and finished — to a recruiter, a hiring manager, or anyone who is not going to run the code themselves.

Screenshots also significantly increase the perceived quality of a project. A clean, well-framed screenshot signals that the builder cares about presentation, not just function.

---

## Requirements

* Every featured project must have at least one screenshot or visual
* The screenshot must show the actual interface or output — not a placeholder or icon
* Screenshots must be clean — no browser chrome unless relevant, no developer tools visible, no test data
* Demo GIFs or video recordings accepted as alternatives for interactive projects
* Live links accepted as primary proof where the deployment is stable

---

## Screenshot Standards

### What to capture

| Project type | What to screenshot |
|---|---|
| Web application | Main dashboard or index view, logged-in state if relevant |
| Data analysis | Key chart or visualisation, summary output |
| CLI tool / script | Terminal output showing the tool running and producing results |
| Desktop app | Main application window in a working state |
| ML project | Model evaluation output, confusion matrix, or prediction sample |

### How to capture

* Use full-window screenshots, not cropped fragments
* Ensure the interface is in a representative, finished state — populated with real or realistic data
* macOS: `Cmd + Shift + 4`, then `Space` to capture a window cleanly
* Linux: Use `scrot` or the GNOME screenshot tool
* For terminal output: increase font size, use a clean theme, capture the relevant section

---

## Upload Workflow

Screenshots are uploaded through the admin panel media manager. No code changes required.

```bash
/admin/projects/<id>/media   # Media manager for each project
```

* **Card image** — uploaded as the thumbnail shown on the projects listing page
* **Screenshots** — one or more images shown on the project detail page
* **Video** — demo recording, shown on detail page with native video player

---

## Audit Checklist

* Every featured project has at least one screenshot uploaded
* Card images are set for all featured projects
* Screenshots show real interface or output — not placeholders
* No developer tools, test data, or draft states visible
* Images are crisp and correctly sized

# 2.4 — Results and Metrics for Data Projects

## Purpose

Data projects require a different kind of proof than web applications. A recruiter hiring for a data analyst role does not just want to see that you wrote pandas code — they want to see that you can derive something meaningful from data and present it clearly.

Results and metrics transform data work from academic exercise into demonstrated analytical capability.

---

## Requirements

Each data and ML project must include at least one of the following:

* A key finding or insight stated in plain language
* A chart, visualisation, or result screenshot showing actual output
* A one-line summary of what the data revealed or what the model achieved
* For ML projects: a specific metric — accuracy, F1 score, RMSE, or equivalent

---

## Results Framework

Each data project description should cover:

```
Finding:  What the analysis revealed — one sentence, specific and concrete.
Method:   How it was done — the tools and approach used.
Output:   What was produced — a chart, a cleaned dataset, a trained model, a report.
```

---

## Examples

**Data analysis project**

Weak:
```
Analysed a dataset using pandas and matplotlib.
```

Strong:
```
Analysed 12 months of sales data to identify a 23% seasonal dip in Q1.
Built visualisations in matplotlib showing category-level trends and provided
a summary report of findings.
```

**Machine learning project**

Weak:
```
Built a classification model with scikit-learn.
```

Strong:
```
Trained a Random Forest classifier on a 10,000-row dataset to predict customer
churn. Achieved 87% accuracy and 0.84 F1 score on the test set. Includes
feature importance analysis and a confusion matrix.
```

---

## Visualisation Upload

Charts and result screenshots are uploaded through the admin panel as screenshots on the project detail page. The same media upload workflow from Section 2.3 applies.

For Jupyter notebook outputs: export the relevant cells as PNG or screenshot the rendered notebook output directly.

---

## Audit Checklist

* Every data or ML project has a key finding stated in the description
* Every data or ML project has at least one result screenshot or chart
* ML projects include a specific performance metric
* Language is concrete and specific — no vague phrases like "gained insights"
* Charts and visualisations are legible at normal viewing size

# Validation

## Recruiter Test

Read the portfolio as if you are a recruiter seeing it for the first time. After sixty seconds, you should be able to answer:

* What does this person build?
* What technologies do they use?
* Is there a link to their code?
* Have I seen something visual that confirms the work is real?
* What did their data projects actually find?

If any of these questions cannot be answered, the relevant section needs further work.

## Non-Technical Reviewer Test

Show the portfolio to someone with no technical background. They should be able to:

* Identify what kind of work is being shown
* See at least one image or visual per featured project
* Click through to at least one live repo or demo
* Understand the purpose of each project from the description alone
