## Phase 1 Architecture Planning




### Objective

Define the portfolio as a system before implementation. Lock decisions that would otherwise cause redesign mid-build.

### What this phase consists of





#### 1. Pages and site structure

1.1 Define core pages: Home, About, Projects, Contact

1.2 Decide optional pages: Project Detail (recommended), Admin/Blog (future)

1.3 Define what content belongs on each page (responsibilities, not HTML yet)

#### 2. Projects hub behavior

2.1 Projects page starts as a hub showing all projects

2.2 Sorting rules: A–Z, newest, oldest

2.3 Filtering rules: primary_category (web/data/software)

2.4 Dedicated category pages exist: /projects/web, /projects/data, /projects/software

2.5 Hybrid classification model: one primary_category + many tags



#### 3. Unified data model (schema)

3.1 Final schema fields (your agreed list):

* id
* title
* primary_category (web / data / software)
* tags (list of strings)
* short_description
* full_description
* tech_stack (list)
* github_link
* live_link (optional)
* image_path
* created_date
* featured (boolean)

3.2 Define sorting fields: title, created_date

3.3 Define filtering fields: primary_category

3.4 Define tag system as informational in Phase 1 but expandable later


#### 4. Navigation flow

4.1 Navbar: Home | About | Projects | Contact

4.2 Define the user journey: landing -> featured -> projects hub -> filters -> detail -> 
GitHub/live

4.3 Decide which pages include “featured projects” components vs full listings




#### 5. Technical stack lock

5.1 Flask + Jinja for the web layer

5.2 Bootstrap + custom CSS for UI

5.3 Database planned (SQLite) but not necessarily used yet

5.4 Deployment target identified (Render or similar)

5.5 Git/GitHub workflow established




---

#### 6. Repo and Folder Structure Plan

6.1 Define top-level repository structure

* Root files (README, requirements, .gitignore)
* `/app` as runnable application
* `/docs` as planning and phase documentation

6.2 Define internal structure of `/app`

* Entry point
* Templates
* Static assets
* Models
* Database folder

6.3 Define purpose and structure of `/docs`

* Phase folders
* Planning notebooks
* Reference maps

6.4 Define deployment alignment

* requirements.txt at root
* Flask entry location
* Production compatibility considerations

6.5 Define `.gitignore` policy

* Environment folders
* Cache files
* Database files
* OS artifacts
* Secret handling principles

---








#### 7. Phase 1 outputs

7.1 Phase 1 spec in docs

7.2 Project classification map (initial entries, expandable)

7.3 A “definition of done” checklist for Phase 1

# Phase 2 — Core Website Build (MVP)



### Objective

Build a functioning website that matches Phase 1 behavior, without adding heavy backend complexity. The site must be runnable, navigable, and complete in structure.



### What this phase consists of


#### 1. Application foundation

1.1 Flask app initialization and config setup

1.2 Route skeleton for core pages

1.3 Environment setup and dependency management

1.4 Confirm server runs locally reliably

#### 2. Template system (Jinja foundation)

2.1 base.html with global layout, navbar, footer

2.2 Template inheritance across all pages

2.3 Shared components: nav, footer, project card partial (optional)

2.4 Clear page-level templates: index/about/projects/contact


#### 3. MVP UI styling

3.1 Bootstrap layout grid

3.2 Typography and spacing baseline

3.3 Simple responsive behavior

3.4 Static assets pipeline: css folder, images, optional JS

#### 4. Projects hub implementation (key feature of Phase 2)

4.1 Use a temporary in-memory dataset (Python list/dict) that follows your schema

4.2 Render project cards via Jinja loop

4.3 Sorting: A–Z, newest, oldest

4.4 Filtering by primary_category

4.5 Dedicated category pages: web/data/software views

4.6 Featured projects on Home pulled from featured=True




#### 5. Project detail pages

5.1 Route: /projects/<id> or /project/<slug>

5.2 Detail template renders full_description, tech_stack, tags, links, image

5.3 Navigation back to Projects hub












#### 7. Phase 2 outputs

7.1 Fully working MVP locally

7.2 Projects hub meets behavior spec

7.3 Readable UI baseline

7.4 No broken routes, no placeholder navigation

# Phase 3 — Database Integration



### Objective

Replace the in-memory project dataset with a persistent data layer while keeping the same schema and UI behavior.



### What this phase consists of


#### 1. Database decision and initialization

1.1 SQLite database file strategy

1.2 Decide whether to use raw SQLite or SQLAlchemy ORM

1.3 Create schema to match the unified project model

1.4 Decide how to store lists (tags/tech_stack): JSON string, join table, or separate tables


#### 2. Project data persistence

2.1 Insert projects into DB (seed script or manual insert)

2.2 Replace in-memory list with DB queries

2.3 Ensure sorting and filtering still work, now DB-driven

#### 3. Contact message persistence

3.1 Contact form writes to DB

3.2 Validation, timestamping

3.3 Optional admin view later







#### 4. Phase 3 outputs

4.1 Project listings come from DB

4.2 Detail pages come from DB

4.3 Contact messages persist

4.4 Schema remains identical to Phase 1 spec

### What this phase consists of



#### 1. Form UX and validation

3.2 Messages are stored and/or emailed

3.3 UX feels professional

# Phase 5 — Data Science Showcase Integration



### Objective

Use your advantage: you’re not only “web”. You’re “web + data”. This phase integrates your data work into the portfolio in a way that is understandable and impressive.



### What this phase consists of



#### 1. Data projects as first-class portfolio entries

1.1 Data projects use the same unified schema (primary_category=data + tags)

1.2 Each data project has visuals, methods, results, and links



#### 2. Visualization strategy

2.1 Decide: static images vs embedded Plotly HTML

2.2 If embedded: where files live, how loaded, performance considerations

2.3 Provide clear narrative: problem -> method -> result



#### 3. Phase 5 outputs

3.1 Data projects appear seamlessly alongside others

3.2 Visuals and writeups are readable

3.3 “Proof of skill” is obvious
