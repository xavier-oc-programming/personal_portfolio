# Phase 9 — Version 2 (Power Mode Upgrades)

## Objective

Optional engineering maturity layer. Not required, but valuable if you want to push beyond a typical junior portfolio.

---

# What this phase consists of

## 1. Admin Dashboard

### 1.1 Login / Auth

Implement session-based authentication to protect the admin panel. Only authenticated users should be able to access any admin route.

### 1.2 Add / Edit / Remove Projects

Full CRUD interface for managing portfolio projects through the UI — no database or codebase access required.

### 1.3 Manage Tags / Categories

A dedicated interface to create, rename, and delete the tags and categories used to filter and organise projects on the public-facing site.

---

## 2. API Layer

### 2.1 REST Endpoints for Projects

Expose project data through a clean REST API. Each project should be retrievable individually or as a collection via structured endpoints.

### 2.2 JSON Responses

All API responses should return well-structured JSON — consistent field naming, appropriate status codes, and predictable error shapes.

### 2.3 Potential Front-End Decoupling Later

The API layer sets the foundation for a future decoupled architecture — allowing the front end to be rebuilt in React or another framework without touching the back end.

---

## 3. DevOps Upgrades

### 3.1 Docker

Containerise the application using Docker. The portfolio should run identically in development and production with a single command.

### 3.2 CI/CD Pipeline

Automate testing and deployment using a CI/CD pipeline (e.g. GitHub Actions). Every push to main should trigger a build and deploy without manual intervention.

### 3.3 Automated Tests

Write a suite of automated tests covering core routes, database operations, and API endpoints. A passing test suite becomes a proof signal in itself.

---

## 4. Phase 9 Outputs

### 4.1 Portfolio Becomes a Maintainable Mini-Platform

By the end of this phase, the portfolio is no longer just a project showcase — it is a small production system with authentication, an API, containerisation, automated tests, and a deployment pipeline.

This demonstrates engineering maturity well beyond a typical junior portfolio and provides concrete talking points for technical interviews.
