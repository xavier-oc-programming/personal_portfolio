---

## Mobile Navbar

The navigation system is refined to ensure proper behavior on mobile devices.

Key improvements include:

* collapsing navigation items into a mobile menu using Bootstrap
* improving spacing for touch interactions
* ensuring the navbar does not overflow on small screens
* supporting clear and reliable menu toggling behavior

The navbar uses:

```html
navbar-expand-lg
```

This ensures that:

* the full navigation is visible on large screens
* the menu collapses into a toggle on smaller devices

Additional refinements improve usability by:

* ensuring links are easy to tap
* maintaining clear visual hierarchy
* preventing layout breaking on narrow viewports

---

## Card Layout Scaling

Project cards are refined to scale smoothly across different screen sizes using the Bootstrap grid system.

Responsive behavior ensures:

* multiple cards per row on large screens
* fewer cards per row on medium screens
* stacked cards on small screens

Example layout progression:

```
Desktop: 3–4 cards per row
Tablet:  2 cards per row
Mobile:  1 card per row
```

This is achieved using responsive column classes such as:

```html
col-12 col-md-6 col-lg-4
```

Additional improvements ensure:

* consistent spacing between cards
* stable alignment regardless of content length
* predictable wrapping behavior for buttons and tags

---

## Image Sizing and Cropping Consistency

Images used across project cards and detail pages are standardized to maintain visual consistency.

Common issues addressed include:

* stretched or distorted images
* inconsistent aspect ratios
* uneven card heights due to variable image sizes

Solutions include:

* fixed height containers for card thumbnails
* consistent border radius and layout integration
* controlled cropping behavior using:

```css
object-fit: cover;
```

This ensures that:

* images fill their containers properly
* proportions remain visually balanced
* layouts remain consistent regardless of source image dimensions

---

## Accessibility Basics

Accessibility improvements are introduced to make the application usable for a wider range of users.

Key practices include:

* meaningful **alt text** for all images
* use of semantic HTML elements (`section`, `nav`, `header`, etc.)
* sufficient color contrast for readability
* proper heading hierarchy across pages
* keyboard-friendly navigation

Example improvement:

```html
<img alt="Screenshot of portfolio project dashboard">
```

Additional refinements ensure:

* form inputs are properly labeled
* navigation links are clearly identifiable
* interactive elements are accessible via keyboard

These improvements enhance both usability and SEO.

---

## Basic JavaScript Enhancements

A lightweight JavaScript layer is introduced to improve interaction without replacing the server-rendered architecture.

JavaScript is used as a **thin enhancement layer** on top of Flask + Jinja.

Typical enhancements include:

* improved mobile navigation interaction (if needed beyond Bootstrap)
* dismissible flash messages (Bootstrap integration)
* optional scroll-to-top behavior
* small UI interaction improvements

The application architecture remains:

```
Flask backend
      ↓
Jinja templates
      ↓
CSS styling
      ↓
Small JavaScript enhancements
```

This ensures that:

* the application remains simple and maintainable
* core functionality does not depend on JavaScript
* user experience is enhanced without added complexity

---

## JavaScript Integration

The project introduces a local JavaScript file:

```bash
app/static/js/main.js
```

This file is connected in the base layout:

```html
<script src="{{ url_for('static', filename='js/main.js') }}"></script>
```

This enables future interaction improvements while keeping all custom behavior centralized.
