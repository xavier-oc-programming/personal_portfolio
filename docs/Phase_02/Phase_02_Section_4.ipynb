# Phase 2 · Section 4 — Projects Hub and Dynamic Project Rendering

### Objective

Phase 2 Section 4 introduces the **Projects Hub**, transforming the portfolio from a collection of static pages into a **data-driven application capable of dynamically rendering project content**.

This section establishes a centralized project dataset, dynamic routing for project listings and detail pages, and reusable UI components for displaying project information consistently across the site.

The goal is to separate **project content, application logic, and presentation**, allowing the portfolio to scale cleanly as additional projects are added.

By the end of this section, the application can dynamically render project cards, filter and sort project listings, and generate dedicated pages for each project using a structured dataset.

---

## Centralized Project Dataset

Project information is now defined in a dedicated module:

```
app/data/projects.py
```

This module acts as an **in-memory data source** for the portfolio during the MVP phase.

Projects are represented as structured dictionaries following a consistent schema that includes:

* project slug
* title
* primary category
* tags
* short description
* full description
* technology stack
* project links
* media assets
* featured status
* project date
* structured narrative fields such as problem, solution, challenges, and results

By centralizing project data in a dedicated module, the application separates **content from presentation**, making the system easier to maintain and extend.

Helper functions are also introduced to access project data cleanly:

```
get_all_projects()
get_project_by_slug(slug)
```

These helpers provide a clear interface for retrieving project information while keeping route logic simple.

---

## Dynamic Projects Hub Route

The `/projects` route is upgraded to dynamically load and render project data from the centralized dataset.

The route performs several responsibilities before passing data to the template:

* retrieving the full project list
* applying optional category filtering
* applying optional sorting
* computing category counts
* passing the processed dataset to the template

This transforms the Projects page from a static layout into a **dynamic project browser** capable of adapting automatically as new projects are added.

---

## Project Filtering

The Projects Hub now supports filtering by primary category.

Supported categories include:

```
web
data
software
```

Filtering is implemented using URL query parameters such as:

```
/projects?category=web
/projects?category=data
/projects?category=software
```

This approach allows the same template and route to serve multiple filtered views of the project dataset while keeping routing logic simple.

The interface visually highlights the active category to provide clear feedback to the user.

---

## Project Sorting

Project listings can now be sorted using query parameters.

Supported sorting options include:

```
A–Z
Newest
Oldest
```

Examples:

```
/projects?sort=az
/projects?sort=newest
/projects?sort=oldest
```

Sorting logic is handled at the route level, ensuring that templates remain focused solely on presentation.

---

## Reusable Project Card Component

A reusable component is introduced to standardize the display of projects within the hub.

```
templates/components/project_card.html
```

Each project card renders:

* project thumbnail image
* project title
* category badge
* short description
* technology stack preview
* project tags
* GitHub repository link
* project detail page link

Using a component ensures visual consistency and makes future layout changes easier to maintain.

---

## Dynamic Project Grid Rendering

The Projects page now renders project cards through a Jinja loop using data passed from the route.

```
{% for project in projects %}
    {% include "components/project_card.html" %}
{% endfor %}
```

This replaces the earlier static placeholder cards and allows the grid to scale automatically as the dataset grows.

---

## Featured Projects Integration

The Home page Featured Projects section now pulls directly from the centralized dataset rather than using placeholder content.

Projects marked with the `featured` flag automatically appear in the Featured section of the Home page.

This ensures that project content remains consistent across the site and eliminates duplicated definitions.

---

## Dynamic Project Detail Pages

Each project now has a dedicated detail page generated dynamically using its unique slug.

```
/projects/<slug>
```

The route retrieves the correct project from the dataset using:

```
get_project_by_slug(slug)
```

The project detail template then renders structured project information including:

* full project description
* category and tags
* technology stack
* project links
* problem statement
* solution description
* development challenges
* results or outcomes

Fallback logic ensures the template remains stable even if certain project fields are not populated.

---

## Project Media System

Projects now support a structured media configuration that separates visual assets by purpose.

Each project can define:

```
media:
    card_image
    screenshots
    videos
```

This allows the portfolio to distinguish between:

* thumbnail images used in project cards
* screenshots displayed on project detail pages
* optional project demonstration videos

The project detail page includes a screenshot gallery where the card image is displayed first, followed by any additional screenshots.

---

## Static Media Organization

Project media assets are organized by project slug within the static directory.

```
static/images/projects/<project-slug>/card/
static/images/projects/<project-slug>/screenshots/
static/videos/projects/<project-slug>/
```

This structure keeps assets grouped by project and prevents naming conflicts as the portfolio grows.

---

## Template Improvements

Several templates are updated to support dynamic project rendering:

```
templates/
    index.html
    projects.html
    project_detail.html
    components/project_card.html
```

Updates include:

* dynamic project rendering via Jinja loops
* reusable UI components
* improved template readability
* conditional rendering for optional project fields

These changes improve maintainability and reduce duplication across templates.

---

## Result

With Section 4 complete, the portfolio application now features a fully functional **Projects Hub** powered by structured project data.

The site can dynamically render project listings, filter and sort projects, display featured projects on the Home page, and generate dedicated pages for each project.

This marks the transition from a static MVP layout to a **data-driven portfolio system**, laying the foundation for future enhancements such as database-backed projects, richer media galleries, and advanced project documentation.
