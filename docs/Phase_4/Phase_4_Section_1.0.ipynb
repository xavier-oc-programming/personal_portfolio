# Phase 4 · Section 1 — Contact Form Validation and Input Sanitization

### Objective

Phase 4 Section 1 strengthens the contact message system introduced in Phase 3 by improving how form submissions are validated and processed.

While Phase 3 enabled the portfolio to store messages submitted through the contact page, validation remained relatively basic. This section introduces structured validation rules, centralized field constraints, and input sanitization to improve data quality and security.

By the end of this section, contact messages are validated, sanitized, and consistently formatted before being stored in the database.

---

## Prerequisites

Before completing this section, the following components must already exist:

* SQLite database configuration
* SQLAlchemy ORM integration
* Contact message database model
* Basic contact form submission route

These components were implemented in **Phase 3 · Section 3 — Contact Message Storage**.

---

## Implementation Steps

The following improvements are introduced in this section:

1. Define centralized field length constraints.
2. Implement structured server-side validation.
3. Introduce reusable input sanitization utilities.
4. Normalize whitespace and escape unsafe characters.
5. Improve validation feedback using flash messages.

These steps create a more reliable and secure input handling pipeline.

---

## Centralized Field Constraints

Contact form limits are defined as constants within the application.

Typical constraints include:

```
MAX_NAME_LENGTH = 60
MAX_EMAIL_LENGTH = 254
MAX_MESSAGE_LENGTH = 2000
```

Defining these limits centrally ensures consistent validation between backend logic and the HTML form.

---

## Server-Side Validation

Server-side validation ensures submitted data meets expected requirements.

Validation checks include:

* required field validation
* minimum character limits
* maximum character limits
* email format verification

Validation occurs before any message object is created or stored.

If validation fails, the form is re-rendered and the user receives clear feedback.

---

## Input Sanitization

All user input is sanitized before being stored.

Two sanitization utilities are used.

### Short-field sanitization

Applied to:

```
name
email
```

This process:

* trims whitespace
* normalizes spacing
* escapes HTML-sensitive characters
* enforces maximum length limits

### Message sanitization

Used for the message field.

This process:

* preserves paragraph breaks
* normalizes whitespace
* escapes HTML-sensitive characters
* enforces message length limits

These steps ensure stored data remains safe and readable.

---

## Validation Feedback

If validation fails, the application provides immediate feedback.

Flash messages communicate issues such as:

* missing required fields
* invalid email format
* message length violations

Previously entered form values are preserved so users do not need to re-enter their message.

---

## Submission Workflow

Once a submission passes validation and sanitization, the message is stored.

Typical workflow:

```
1. Receive POST request
2. Extract form values
3. Validate input
4. Sanitize values
5. Create ContactMessage object
6. Insert record using SQLAlchemy
7. Commit transaction
8. Redirect user to contact page
```

This ensures only validated and sanitized messages are stored.
