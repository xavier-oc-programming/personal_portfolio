# Phase 4 · Section 1.5 — Structured Form Handling and Security with Flask-WTF

### Objective

Phase 4 Section 1.5 further strengthens the contact message system by introducing **Flask-WTF and WTForms** for structured form handling and additional security protections.

While Phase 4 Section 1 implemented manual validation and sanitization logic, the form was still processed directly through `request.form`. This section replaces that manual form structure with a dedicated **form class** that centralizes validation logic and enables automatic CSRF protection.

By the end of this section, the contact form uses a **Flask-WTF form class**, providing a more maintainable, secure, and scalable approach to form validation and submission handling.

* SQLite database configuration

The following improvements are introduced in this section:

## Flask-WTF Form Integration

Flask-WTF integrates WTForms with Flask and provides a structured way to define and validate forms.

The contact system now uses a dedicated form class instead of manually parsing request data.

Typical usage pattern:

```
form = ContactForm()

if form.validate_on_submit():
    ...
```

This method automatically checks:

* whether the request is a valid POST submission
* whether all field validation rules pass

If validation fails, the form is re-rendered with validation errors.

## Contact Form Class

A dedicated form class is introduced within the application.

Typical fields include:

```
name
email
message
honeypot
```

Each field defines its own validation rules using WTForms validators.

Example validators include:

* `DataRequired`
* `Email`
* `Length`

Centralizing validation inside the form class keeps route logic clean and improves maintainability.

---

## CSRF Protection

Flask-WTF automatically protects forms against **Cross-Site Request Forgery (CSRF)** attacks.

Each form submission includes a hidden token generated by the server.

Example template integration:

```
{{ form.hidden_tag() }}
```

If the CSRF token is missing or invalid, the submission is rejected.

This prevents malicious websites from submitting requests on behalf of legitimate users.

---

## Honeypot Spam Detection

A lightweight spam protection mechanism is introduced using a **honeypot field**.

Example field:

```
company
```

This field is hidden from normal users but visible to automated bots that attempt to fill all form inputs.

If the honeypot field contains a value during submission, the request is treated as spam and rejected.

This provides basic bot protection without requiring external services.

---

## Submission Timing Protection

An additional anti-spam measure is implemented using **submission timing checks**.

When the contact page loads, the application records a timestamp in the user session.

Example:

```
session["contact_form_loaded_at"]
```

When the form is submitted, the application verifies that a minimum amount of time has passed.

Example rule:

```
MIN_FORM_FILL_SECONDS = 3
```

Submissions that occur faster than this threshold are rejected as likely automated requests.

---

## Rate Limiting

Basic rate limiting is introduced to prevent repeated rapid submissions.

The application stores the timestamp of the last successful contact message:

```
session["last_contact_submission_at"]
```

A cooldown window prevents multiple submissions within a short period.

Example rule:

```
CONTACT_RATE_LIMIT_SECONDS = 60
```

If another submission occurs within this window, the request is rejected.

---

## Submission Workflow

Once a submission passes validation, spam checks, and sanitization, the message is stored.

Typical workflow:

```
1. Receive POST request
2. Validate form using Flask-WTF
3. Perform honeypot spam check
4. Verify submission timing
5. Enforce rate limiting
6. Sanitize input values
7. Create ContactMessage object
8. Insert record using SQLAlchemy
9. Commit transaction
10. Redirect user to contact page
```

This ensures only legitimate and properly validated messages are stored.
