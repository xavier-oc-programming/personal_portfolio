## Input Normalization

Before messages are stored, all form inputs are normalized to maintain consistent formatting.

Normalization includes:

* trimming leading and trailing whitespace
* collapsing excessive internal whitespace
* preserving intentional line breaks in messages

This ensures that stored messages remain clean and readable.

---

## Safe Database Persistence

Validated submissions are stored using SQLAlchemy.

Typical pattern:

```
db.session.add(contact_message)
db.session.commit()
```

If an error occurs during commit, the session is rolled back to maintain database stability.

```
db.session.rollback()
```

This prevents corrupted ORM sessions and ensures reliable persistence.

---

## Email Notification

An optional notification email is sent after a message is successfully stored.

The system uses **Flask-Mail**, which integrates SMTP email services with Flask.

Typical configuration includes:

```
MAIL_SERVER
MAIL_PORT
MAIL_USERNAME
MAIL_PASSWORD
MAIL_USE_TLS
MAIL_DEFAULT_SENDER
CONTACT_NOTIFICATION_EMAIL
```

For development, Gmail SMTP can be used with an **App Password**.

---

## Submission Workflow

A successful contact submission follows this sequence:

```
1. Receive POST request
2. Validate form using Flask-WTF
3. Perform spam checks
4. Normalize input values
5. Create ContactMessage object
6. Insert record using SQLAlchemy
7. Commit transaction
8. Send notification email
9. Redirect user to contact page
```

Database persistence occurs **before email delivery** to ensure messages are never lost.

---

## Spam Protection

Several protections help prevent automated abuse:

* honeypot field detection
* submission timing validation
* rate limiting repeated submissions
* CSRF protection
* server-side input validation

These mechanisms reduce automated spam while allowing legitimate submissions.

---

## Activity Logging

Spam-related events are recorded in the application logs.

Examples include:

```
Contact form honeypot triggered
Contact form submitted too quickly
Contact form rate limit triggered
```

Logging assists with debugging and helps confirm that spam protections are functioning correctly.
