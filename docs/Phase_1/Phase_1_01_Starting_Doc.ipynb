## 2. Projects Hub Behavior

This section defines how projects are displayed, organized, and navigated within the portfolio.

It establishes the rules for default 
- presentation, 
- sorting, 
- filtering,
- category separation,
- and hybrid classification

to ensure structural clarity and scalable growth.

--- 

### 2.1 Projects page starts as a hub showing all projects

#### Default View

* Displays all projects
* Uses a responsive grid layout
* Default sort: Newest first

---

### 2.2 Sorting rules: A–Z, newest, oldest

#### Sorting Options

User can sort by:

* Alphabetical (A–Z)
* Newest first
* Oldest first

Sorting applies after filters.

---

### 2.3 Filtering rules: primary_category (web/data/software)

#### Category Filtering

Filter buttons:

* All
* Web
* Data
* Software

Filtering uses primary_category only.

---

### 2.4 Dedicated category pages exist

#### Dedicated Category Pages

Separate pages exist:

* /projects/web
* /projects/data
* /projects/software

Each page:

* Shows only projects with that primary_category
* Still supports sorting
* Displays tags per project

These reuse the same template logic.

---

### 2.5 Hybrid classification model: one primary_category + many tags

Projects are organized using:

* One **primary_category**
* Multiple **tags**

Primary category is determined by the project’s primary delivery context:

* **Web** — Main experience delivered via browser (Flask app, routes, templates), even if it includes ML or data.
* **Data** — Core deliverable is analysis, modeling, visualization, or dashboards.
* **Software** — Core deliverable is a CLI, GUI, automation tool, or non-web application logic.

Tie-breaker rule:

If ambiguous, assign the category that reflects:

1. The main user interface
2. Where most engineering complexity resides
3. What defines the project’s identity if other layers are removed

Tags preserve hybrid depth without affecting structural placement.

## 3. Unified Data Model (Schema) — For Projects in Portfolio

This section defines the standardized structure used to represent every project in the portfolio.

The schema ensures consistency across pages, supports sorting and filtering behavior, <br> and provides a scalable foundation for future database integration.

---

### 3.1 Final schema fields

Final schema:

* id
* title
* primary_category (web / data / software)
* tags (list of strings)
* short_description
* full_description
* tech_stack (list)
* github_link
* live_link (optional)
* image_path
* created_date
* featured (boolean)

This schema supports sorting, filtering, tagging, and future scalability.

---

### 3.2 Sorting fields

Sorting is based on:

* title
* created_date

Default sort:

* created_date (Newest first)

---

### 3.3 Filtering fields

Filtering is based on:

* primary_category

Tag-based filtering is not active in Phase 1.

---

### 3.4 Tag system rules

Tags:

* Stored as a list of strings
* Used for display and skill signaling
* Do not affect structural placement
* Do not override primary_category

Constraints:

* Each project must have exactly one primary_category
* Each project may have multiple tags
* Tags should represent technologies, concepts, or techniques (e.g., flask, pandas, regression, selenium)

Tags preserve hybrid accuracy without compromising structural clarity.

## 5. Technical Stack Lock

This section locks the core technology choices for Version 1. The goal is to prevent tooling drift mid-build and ensure the portfolio is implemented with a consistent, production-oriented stack that supports future scalability.

---

### 5.1 Web Layer: Flask + Jinja

* Backend framework: Flask
* Templating: Jinja2
* Rendering approach: server-side rendered pages (SSR)
* Routing structure aligns with the core pages and project detail pages
* Templates organized for reuse (base layout + page templates + partials as needed)

This stack supports fast iteration, clear separation of concerns, and deployable production behavior.

---

### 5.2 UI Layer: Bootstrap + Custom CSS

* UI framework: Bootstrap (layout, grid, base components)
* Custom CSS used for:

  * Typography system
  * Spacing scale
  * Button styling
  * Brand-level polish beyond default Bootstrap

Bootstrap provides speed and responsiveness; custom CSS ensures identity and consistency.

---

### 5.3 Data Layer Planning: SQLite (Phase 1 Planning Only)

* Database target: SQLite
* Phase 1 scope:

  * Schema is defined
  * Database integration is not implemented yet
* Data source in early implementation may be static (e.g., JSON or Python dicts), but must conform to the unified schema

SQLite is chosen for simplicity, portability, and easy upgrade path to a stronger database later if needed.

---

### 5.4 Deployment Target

* Deployment target: Render (or equivalent production hosting)
* Requirements:

  * WSGI-compatible deployment
  * Environment variables supported
  * `requirements.txt` driven installs
  * Git-based deploy from repository

Final provider can be confirmed later, but deployment constraints are locked now to avoid local-only design.

---

### 5.5 Git/GitHub Workflow Structure

* Repository is the single source of truth
* Branching:

  * `main` remains stable and deployable
  * Feature branches per phase or per feature set
* Commit style:

  * Small, descriptive commits tied to doc and build steps
* Documentation is stored alongside code (planning and build traceability)

This ensures clean iteration and professional workflow visibility.

---

### 5.6 Environment Management Decision (venv vs conda)

* Preferred runtime environment for the portfolio repo: `venv`
* Rationale:

  * Most production hosts expect pip + `requirements.txt`
  * Minimizes dependency ambiguity
  * Keeps deployment aligned with local development

Conda environments may still exist globally on the machine, but the portfolio project itself should be reproducible via:

* `python -m venv .venv`
* `pip install -r requirements.txt`

---

### 5.7 Development vs Production Configuration Separation

Configuration is split to avoid hardcoding secrets or environment-specific settings.

* Development:

  * Local debug mode enabled
  * Local database path (if/when used)
  * Local environment variables via `.env` (ignored by git)

* Production:

  * Debug disabled
  * Environment variables injected by hosting provider
  * Secure secret key handling
  * Production-ready server (WSGI entrypoint)

Config principles:

* No secrets committed to repo
* Environment variables drive sensitive values
* Settings allow local execution and production deployment without code changes
