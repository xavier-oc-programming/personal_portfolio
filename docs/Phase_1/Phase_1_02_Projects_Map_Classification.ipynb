# Project Classification Map (Phase 1 Scope)

## Schema Used (Phase 1 Version)

For Phase 1, we define only the fields required to support:

* Categorization
* Sorting
* Filtering

We are not fully populating the final database schema yet. We are locking structure, not content depth.

Each project entry includes:

* title
* primary_category (web / data / software)
* tags (list of strings)
* created_date
* github_link

Sorting fields (already supported conceptually):

* title
* created_date

Filtering field:

* primary_category

Tags:

* Informational in Phase 1
* Expandable later
* Used to reflect hybrid skill usage
