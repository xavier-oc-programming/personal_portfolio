# Phase 7 · Section 2 — Hosting and Deployment

### Objective

Phase 7 Section 2 transitions the portfolio application from a **production-ready codebase** into a **live, publicly accessible web application**.

Following the configuration and security setup in Section 1, this section focuses on **deploying the application to a cloud hosting platform**, ensuring it runs reliably in a real-world environment.

By the end of this section, the portfolio is **fully deployed, accessible via a public URL, and functioning in a production environment**.

---

## Prerequisites

The following components must already exist:

* Environment-based configuration (`config.py`)
* Secure `.env` handling and `.env.example` template
* WSGI entrypoint (`wsgi.py`)
* Dependency management via `requirements.txt`
* Fully functional Flask application (routes, database, UI, contact system)

These were implemented in:

* **Phase 7 · Section 1 — Production Readiness and Configuration**
* **Phase 6 — UI Refinement and Performance Optimization**
* **Phase 4 — Contact System (Professional Functionality)**

## Implementation Steps

This section introduces deployment through a **structured and reproducible workflow**:

1. Prepare deployment files (`requirements.txt`, `wsgi.py`, `.env.example`)
2. Connect the GitHub repository to a hosting platform
3. Configure build and start commands
4. Set environment variables in the hosting dashboard
5. Deploy the application
6. Verify the live application behavior

---

## Deployment File Preparation

### requirements.txt

All dependencies required to run the application are defined:

```bash
pip install -r requirements.txt
```

Includes:

* Flask
* Flask-SQLAlchemy
* Flask-WTF
* Flask-Mail
* python-dotenv
* Gunicorn-compatible stack

This ensures the production environment matches local development.

---

### WSGI Entry (wsgi.py)

A dedicated WSGI entrypoint is provided:

```python
from app.app import create_app

app = create_app()
```

This allows the application to be executed by production servers such as Gunicorn.

---

### Environment Template (.env.example)

A template file is provided:

```bash
SECRET_KEY=your-secret-key-here
MAIL_USERNAME=your-email@example.com
MAIL_PASSWORD=your-app-password
```

This ensures:

* safe onboarding for new environments
* no secrets committed to version control
* consistent configuration structure

---

## Hosting Platform Setup

The application is deployed using a cloud platform such as:

* **Render**
* **Railway**

Typical setup:

* Connect GitHub repository
* Create a new Web Service
* Select Python environment
* Choose deployment branch (e.g., `main`)

---

## Build and Start Commands

### Build Command

```bash
pip install -r requirements.txt
```

### Start Command

```bash
gunicorn wsgi:app
```

This ensures:

* proper dependency installation
* execution via a production-ready WSGI server

---

## Environment Variables Configuration

Environment variables are configured directly in the hosting platform.

Example:

```bash
FLASK_ENV=production
SECRET_KEY=...

MAIL_SERVER=smtp.gmail.com
MAIL_PORT=587
MAIL_USE_TLS=true

MAIL_USERNAME=...
MAIL_PASSWORD=...
CONTACT_NOTIFICATION_EMAIL=...
```

This ensures:

* secrets are not exposed in code
* configuration is environment-specific
* production remains secure and flexible

---

## Deployment Execution

Once configured, the platform:

1. Clones the repository
2. Installs dependencies
3. Starts the application via Gunicorn

A public URL is generated:

```
https://your-app-name.onrender.com
```

---

## Live Application Verification

After deployment, the application is tested via the public URL.

Key checks:

* Homepage loads correctly
* Navigation routes function properly
* Project pages render from database
* Static assets (CSS, JS, images) load correctly
* Contact form:

  * validates input
  * stores messages in database
  * sends email notifications
* No runtime errors appear

---

## Deployment Considerations

Common issues addressed:

* Missing environment variables
* Incorrect start command
* Static file path issues
* Database file initialization (`_ensure_database_directory_and_file`)
* Email service configuration

These checks ensure stable production behavior.
