## DNS Configuration

Domain records are configured in the registrar’s DNS panel.

### CNAME Record

Maps the domain to the hosting provider:

```txt
Type: CNAME
Host: www
Value: your-app.up.railway.app
```

This ensures that traffic to the custom domain is routed to the deployed application.

---

### TXT Record (Verification)

Used by the hosting provider to verify ownership:

```txt
Type: TXT
Host: _railway-verify.www
Value: railway-verify=xxxxxxxx
```

This confirms domain control before activation.

---

### Root Domain Redirect

Optional but recommended:

```txt
Type: URL Redirect
Host: @
Value: http://www.yourdomain.com
```

This ensures consistent routing between root and `www` versions.

---

* Confirm domain activation

```
https://www.xavieroc.dev
```

## HTTPS Configuration

HTTPS is automatically enabled by the hosting provider using SSL certificates.

### Key aspects:

* SSL certificate provisioning (automatic)
* Secure connection via HTTPS
* Encryption of all client-server communication

Example:

```
https://www.xavieroc.dev
```

---

## HTTP → HTTPS Redirection

The application should ensure all traffic is served securely.

This is typically handled by the hosting platform and includes:

* automatic redirect from HTTP to HTTPS
* prevention of insecure access

---

## Live Domain Verification

After configuration, the domain must be tested:

### Key checks:

* Domain resolves correctly
* HTTPS loads without warnings (🔒 visible in browser)
* All routes function properly
* Static assets load correctly
* Forms and backend logic operate normally

---

## Deployment Considerations

Common issues addressed:

* DNS propagation delays
* Incorrect CNAME values
* Conflicting DNS records (e.g., parking pages)
* Missing TXT verification
* SSL certificate provisioning delay

Proper configuration ensures stable and secure domain access.
