# Phase 7 — Deployment (Production Release)

## Objective

Make the portfolio application **live, accessible, and production-ready**.

The goal of this phase is to transition the application from a local development environment to a **publicly accessible production deployment**, ensuring reliability, security, and proper configuration.

---

# What this phase consists of

## 1. Production readiness

### 1.1 Environment variables

Sensitive configuration must be moved out of the codebase and into environment variables.

Typical variables include:

* `SECRET_KEY`
* `DATABASE_URL`
* `EMAIL_API_KEY` (if applicable)
* `FLASK_ENV`

This ensures that sensitive data is not exposed in version control.

---

### 1.2 Production config vs development

The application must support separate configurations for development and production.

Production configuration should:

* Disable debug mode
* Use secure settings
* Load environment variables instead of hardcoded values

This separation ensures safe and predictable behavior in production.

---

### 1.3 Debug disabled

Debug mode must be turned off in production:

```
DEBUG = False
```

This prevents:

* exposure of internal errors
* security vulnerabilities
* unintended behavior

---

### 1.4 Proper WSGI entry

The application must expose a proper WSGI entrypoint compatible with hosting providers.

Example:

```
gunicorn app.app:app
```

This ensures the app can be served by a production-ready web server.

---

## 2. Hosting

### 2.1 Deployment platform setup (Render / Railway)

The application is deployed to a hosting provider such as:

* Render
* Railway

Typical setup includes:

* connecting the GitHub repository
* selecting the Python environment
* defining build and start commands

---

### 2.2 Build and install commands

The hosting platform must install dependencies and prepare the app.

Typical commands:

```
pip install -r requirements.txt
```

Start command:

```
gunicorn app.app:app
```

This ensures the application runs consistently in production.

---

### 2.3 Live URL verification

After deployment, the application must be tested through its public URL.

Checks include:

* homepage loads correctly
* navigation works
* project pages render
* contact form functions properly

This confirms the deployment is successful.

---

## 3. Optional domain

### 3.1 Custom domain mapping

A custom domain can be connected to the deployed application.

Example:

```
www.yourportfolio.com
```

This improves professionalism and branding.

---

### 3.2 HTTPS verification

HTTPS must be enabled to ensure secure communication.

This includes:

* SSL certificate activation (usually automatic via hosting provider)
* redirecting HTTP → HTTPS

This ensures user data is transmitted securely.

---
