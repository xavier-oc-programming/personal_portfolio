# Analytics Integration

## Purpose

Analytics enables the portfolio to function as a **data-driven application**, providing insights into:

* number of visitors
* most viewed pages
* user navigation behavior
* engagement with projects

This allows continuous improvement based on real usage.

---

## Tool Selection

Analytics is implemented using:

* Google Analytics 4 (GA4)

This is the industry-standard solution for web tracking.

---

## Implementation

Analytics is integrated globally through the base template.

### Location:

```bash
app/templates/base.html
```

### Script (inside `<head>`):

```html
<!-- Google Analytics -->
<script async src="https://www.googletagmanager.com/gtag/js?id=G-XXXXXXXXXX"></script>
<script>
window.dataLayer = window.dataLayer || [];
function gtag(){dataLayer.push(arguments);}
gtag('js', new Date());
gtag('config', 'G-XXXXXXXXXX');
</script>
```

This ensures:

* tracking across all pages
* minimal performance impact
* centralized configuration

# SEO Foundations

## Purpose

SEO ensures the portfolio:

* appears in search engine results
* displays correctly when shared
* communicates structure clearly to search engines

---

## Title Tags (Dynamic)

Each page defines a meaningful title:

```html
<title>{{ page_title or "Xavier OC | Python Developer Portfolio" }}</title>
```

This ensures:

* consistent branding
* improved search ranking
* clear page identification

---

## Meta Description

Add inside `<head>`:

```html
<meta name="description" content="Portfolio of Xavier OC — Python Developer specializing in Flask, Data Analysis, and Automation.">
```

This improves:

* search result previews
* click-through rate

---

## Open Graph Tags

Used for social media previews:

```html
<meta property="og:title" content="Xavier OC Portfolio">
<meta property="og:description" content="Python Developer Portfolio — Flask, Data, Automation">
<meta property="og:type" content="website">
<meta property="og:url" content="https://www.xavieroc.dev">
```

This ensures proper display on:

* LinkedIn
* WhatsApp
* Twitter

---

## Semantic HTML Structure

Templates should follow a clear hierarchy:

* One `<h1>` per page
* Logical heading order (`h1 → h2 → h3`)
* Use of semantic tags:

```html
<header>
<nav>
<main>
<section>
<article>
<footer>
```

This improves:

* accessibility
* SEO ranking
* maintainability

---

## URL Structure

The application already uses clean routes:

```bash
/projects
/projects/<slug>
```

These are:

* human-readable
* SEO-friendly
* scalable
