# Analytics Integration

## Purpose

Analytics enables the portfolio to function as a **data-informed application**, providing insights into:

* number of visitors
* most viewed pages
* user navigation behavior
* engagement with projects
* contact form outcomes
* clicks on portfolio call-to-actions

This allows future iteration based on real user behavior rather than guesswork.

---

## Tool Selection

Analytics is implemented using:

* **Google Analytics 4 (GA4)**

This is a widely used industry-standard solution for modern web tracking.

---

## Configuration Strategy

Analytics is configured through environment variables rather than hardcoded values.

Example:

```bash
GOOGLE_ANALYTICS_ID=G-XXXXXXXXXX
```

This ensures:

* environment-specific configuration
* no hardcoded tracking IDs in templates
* easy deployment updates through the hosting dashboard

---

## Global Integration

Analytics is integrated globally through the base template.

### Core file:

```bash
app/templates/base.html
```

### Behavior

The Google Analytics script only loads if a tracking ID is configured.

Example structure:

```html
{% if google_analytics_id %}
<script async src="https://www.googletagmanager.com/gtag/js?id={{ google_analytics_id }}"></script>
<script>
  window.dataLayer = window.dataLayer || [];
  function gtag() {
    dataLayer.push(arguments);
  }
  gtag("js", new Date());
  gtag("config", "{{ google_analytics_id }}");
</script>
{% endif %}
```

This ensures:

* tracking across all pages
* centralized implementation
* clean separation between code and configuration
* no analytics script in environments where the ID is absent

# SEO Foundations

## Purpose

SEO ensures the portfolio:

* appears more cleanly in search engine results
* displays correctly when shared externally
* communicates page structure clearly to search engines
* maintains consistent metadata across routes

---

## Metadata Configuration

Global SEO-related metadata is configured through application settings and context injection.

Examples include:

```bash
SITE_URL=https://www.xavieroc.dev
SITE_NAME=Xavier OC | Dev Portfolio
DEFAULT_META_DESCRIPTION=...
```

These values are exposed to templates through the Flask application context.

This ensures:

* consistent branding
* centralized metadata management
* reusable SEO defaults across all templates

---

## Title Tags (Dynamic)

Each page now supports dynamic page titles through template blocks.

Example pattern:

```html
<title>{% block title %}{{ site_name }}{% endblock %}</title>
```

This ensures:

* consistent branding
* improved search result clarity
* meaningful page-specific titles

Examples include:

* home page title from `SITE_NAME`
* project pages using the project title
* dedicated titles for About, Projects, and Contact

---

## Meta Description

Each page now supports a dynamic or default meta description.

Example pattern:

```html
<meta name="description" content="{% block meta_description %}{{ default_meta_description }}{% endblock %}">
```

This improves:

* search result previews
* click-through rate
* consistency across the site

---

## Canonical URLs

Canonical tags are included to communicate the preferred version of each page URL.

Example pattern:

```html
<link rel="canonical" href="{% block canonical_href %}{{ canonical_url }}{% endblock %}">
```

This helps reduce ambiguity for search engines and supports cleaner indexing.

---

## Open Graph Tags

Open Graph metadata is included for improved link previews when the portfolio is shared externally.

Example fields include:

```html
<meta property="og:title" content="...">
<meta property="og:description" content="...">
<meta property="og:type" content="website">
<meta property="og:url" content="...">
<meta property="og:image" content="...">
```

This improves presentation on platforms such as:

* LinkedIn
* WhatsApp
* Discord
* other messaging or social platforms

---

## Open Graph Image

A default Open Graph image is configured so shared links display a visual preview.

Example asset location:

```bash
app/static/images/og/default-og-image.png
```

The application builds the full public image URL using `SITE_URL`, ensuring that social platforms can access the image properly.

---

## Semantic HTML Structure

Templates continue to follow a clear hierarchy with:

* one primary `<h1>` per page

This improves:

* accessibility

These remain:
